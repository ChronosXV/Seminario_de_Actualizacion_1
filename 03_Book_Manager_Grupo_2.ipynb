````python
# @title Configuración del entorno
RUTA_SRC = "/content/Seminario_de_Actualizacion_1/book_manager/src"

# 1. Configurar la ruta del proyecto
if RUTA_SRC in sys.path:
    sys.path.remove(RUTA_SRC)

sys.path.insert(0, RUTA_SRC)

# 2. Recuperar DATABASE_URL desde Colab
database_url = userdata.get("DATABASE_URL")

if not database_url:
    raise ValueError(
        "No se encontró DATABASE_URL en los secretos de Colab."
    )

os.environ["DATABASE_URL"] = database_url

# 3. Limpiar versiones anteriores de los módulos
for modulo in list(sys.modules):
    if modulo == "book_manager" or modulo.startswith("book_manager."):
        del sys.modules[modulo]

importlib.invalidate_caches()

# 4. Verificar el motor configurado
from book_manager.database.connection import ConexionDB

conexion = ConexionDB()

print("=== CONFIGURACIÓN DEL ENTORNO ===")
print("Ruta del proyecto:", RUTA_SRC)
print("Motor de base de datos:", conexion.engine.dialect.name)

if conexion.engine.dialect.name != "postgresql":
    raise RuntimeError(
        "La conexión no está utilizando PostgreSQL/Supabase."
    )

conexion.engine.dispose()

print("\nEntorno configurado correctamente.")
````

## Muestra del código fuente


````python
from IPython.display import display, HTML, Code
import os
display(HTML("""
<style>
  .jp-RenderedText pre { background-color: #1e1e1e !important; color: #d4d4d4 !important; }
</style>
"""))

!rm -rf /content/sample_data .config
def visualizar_archivos_dark(directorio):
    for raiz, _, archivos in os.walk(directorio):
      if raiz.lower().endswith(('.git','.dvc', '__pycache__')):
        continue
      for nombre in archivos:
        if nombre.lower().endswith(('.py')):
          ruta = os.path.join(raiz, nombre)
          print(f"📁 {ruta}")
          with open(ruta, 'r') as f:
              contenido = f.read()
          display(Code(contenido, language=nombre.split('.')[-1]))
        if nombre.lower().endswith(('.md', '.csv', '.env', '.sql', '.txt', '.json')):
          ruta = os.path.join(raiz, nombre)
          print(f"📁 {ruta}")
          with open(ruta, 'r') as f:
              #contenido = f.read()
              contenido = f.readlines()[:10]
          print("".join(contenido))
          print("\n\n")

visualizar_archivos_dark('/content/')
````
