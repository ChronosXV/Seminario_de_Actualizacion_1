## Muestra del código fuente


````python
from IPython.display import display, HTML, Code
import os

display(HTML("""
<style>
    .jp-RenderedText pre { background-color: #1e1e1e !important; color: #d4d4d4 !important; }
</style>
"""))
!rm -rf /content/sample_data .config
def visualizar_archivos_dark(directorio):
    for raiz, _, archivos in os.walk(directorio):
      if raiz.lower().endswith(('.git','.dvc', '__pycache__')):
        continue
      for nombre in archivos:
        if nombre.lower().endswith(('.py')):
          ruta = os.path.join(raiz, nombre)
          print(f"📁 {ruta}")
          with open(ruta, 'r') as f:
              contenido = f.read()
          display(Code(contenido, language=nombre.split('.')[-1]))
        if nombre.lower().endswith(('.md', '.csv', '.env', '.sql', '.txt', '.json')):
          ruta = os.path.join(raiz, nombre)
          print(f"📁 {ruta}")
          with open(ruta, 'r') as f:
              contenido = f.read()
          print("".join(contenido))
          print("\n\n")

visualizar_archivos_dark('/content/')
````
