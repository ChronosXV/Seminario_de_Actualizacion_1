### 02 Conexion de BD

````python
%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/database/connection.py

import os
from contextlib import contextmanager

from sqlalchemy import create_engine
from sqlalchemy.orm import sessionmaker


class ConexionDB:
    """
    Administra la conexión y las transacciones de la base de datos.

    Utiliza DATABASE_URL cuando está configurada.
    En caso contrario, utiliza SQLite como alternativa local.
    """

    def __init__(self, url_db=None):
        """
        Inicializa la conexión a la base de datos.

        Args:
            url_db (str | None): URL explícita de conexión.
                Si no se proporciona, utiliza DATABASE_URL
                o SQLite como alternativa.
        """
        self.url_db = (
            url_db
            or os.getenv("DATABASE_URL")
            or "sqlite:///book_manager.db"
        )

        self.engine = create_engine(
            self.url_db,
            echo=False
        )

        self.Session = sessionmaker(
            bind=self.engine,
            autoflush=False,
            expire_on_commit=False
        )

    def crear_sesion(self):
        """
        Crea y devuelve una nueva sesión de SQLAlchemy.
        """
        return self.Session()

    @contextmanager
    def obtener_sesion(self):
        """
        Administra una transacción con commit, rollback
        y cierre automático de sesión.
        """
        sesion = self.crear_sesion()

        try:
            yield sesion
            sesion.commit()

        except Exception:
            sesion.rollback()
            raise

        finally:
            sesion.close()
````

### 03 Creacion del Contex Manager

````python
# Modificacion de connection y creacion de Context manager


%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/database/connection.py
"""
Módulo encargado de la conexión y gestión de transacciones.

Utiliza SQLAlchemy para configurar la conexión a la base
de datos y administrar sesiones mediante un Context Manager.
"""

from contextlib import contextmanager

from sqlalchemy import create_engine
from sqlalchemy.orm import sessionmaker


class ConexionDB:
    """
    Administra la conexión y las transacciones de la base de datos.

    Permite configurar el motor de SQLAlchemy, crear sesiones
    y administrar automáticamente las transacciones.

    Attributes:
        url_db (str): URL de conexión a la base de datos.
        engine (Engine): Motor de conexión de SQLAlchemy.
        Session (sessionmaker): Fábrica de sesiones.
    """

    def __init__(self, url_db="sqlite:///book_manager.db"):
        """
        Inicializa la configuración de conexión a la base de datos.

        Args:
            url_db (str): URL de conexión. Por defecto utiliza
                una base de datos SQLite local.
        """
        self.url_db = url_db

        self.engine = create_engine(
            self.url_db,
            echo=False
        )

        self.Session = sessionmaker(
            bind=self.engine,
            autoflush=False,
            expire_on_commit=False
        )

    def crear_sesion(self):
        """
        Crea y devuelve una nueva sesión de SQLAlchemy.

        Returns:
            Session: Nueva sesión asociada al motor de conexión.
        """
        return self.Session()

    @contextmanager
    def obtener_sesion(self):
        """
        Gestiona automáticamente una sesión y su transacción.

        Confirma los cambios mediante commit si las operaciones
        finalizan correctamente. Si ocurre un error, ejecuta
        rollback para revertir la transacción.

        La sesión se cierra siempre al finalizar el bloque.

        Yields:
            Session: Sesión activa para operar con la base de datos.

        Raises:
            Exception: Propaga cualquier error ocurrido durante
                las operaciones de la transacción.
        """
        sesion = self.crear_sesion()

        try:
            yield sesion
            sesion.commit()

        except Exception:
            sesion.rollback()
            raise

        finally:
            sesion.close()
````

### 04 Creacion de tabla en Base de datos

````python
# Creacion de modelos
%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/models/models.py
"""
Modelos ORM del sistema Book Manager.

Define las tablas, sus tipos de datos, claves primarias,
claves foráneas y relaciones mediante SQLAlchemy.
"""

from datetime import date
from decimal import Decimal

from sqlalchemy import (
    CheckConstraint,
    Date,
    ForeignKey,
    Integer,
    Numeric,
    String,
    UniqueConstraint,
)
from sqlalchemy.orm import (
    DeclarativeBase,
    Mapped,
    mapped_column,
    relationship,
)


class Base(DeclarativeBase):
    """Clase base declarativa de los modelos ORM."""
    pass


class GeneroModel(Base):
    """Representa los géneros literarios de los libros."""

    __tablename__ = "generos"

    id: Mapped[int] = mapped_column(Integer, primary_key=True)
    nombre: Mapped[str] = mapped_column(String(100), nullable=False)

    libros: Mapped[list["LibroModel"]] = relationship(
        back_populates="genero"
    )


class EditorialModel(Base):
    """Representa las editoriales de los libros."""

    __tablename__ = "editoriales"

    id: Mapped[int] = mapped_column(Integer, primary_key=True)
    nombre: Mapped[str] = mapped_column(String(150), nullable=False)

    libros: Mapped[list["LibroModel"]] = relationship(
        back_populates="editorial"
    )


class MonedaModel(Base):
    """Representa las monedas utilizadas por el sistema."""

    __tablename__ = "monedas"

    id: Mapped[int] = mapped_column(Integer, primary_key=True)
    codigo: Mapped[str] = mapped_column(
        String(10), nullable=False, unique=True
    )
    nombre: Mapped[str] = mapped_column(String(100), nullable=False)
    simbolo: Mapped[str] = mapped_column(String(10), nullable=False)

    precios: Mapped[list["PrecioModel"]] = relationship(
        back_populates="moneda"
    )

    cotizaciones: Mapped[list["CotizacionDolarModel"]] = relationship(
        back_populates="moneda"
    )


class TipoCotizacionModel(Base):
    """Representa los tipos de cotización del dólar."""

    __tablename__ = "tipos_cotizacion"

    id: Mapped[int] = mapped_column(Integer, primary_key=True)
    nombre: Mapped[str] = mapped_column(String(100), nullable=False)

    cotizaciones: Mapped[list["CotizacionDolarModel"]] = relationship(
        back_populates="tipo_cotizacion"
    )


class LibroModel(Base):
    """Representa los libros y sus relaciones con género y editorial."""

    __tablename__ = "libros"

    id: Mapped[int] = mapped_column(Integer, primary_key=True)
    isbn: Mapped[str] = mapped_column(
        String(20), nullable=False, unique=True
    )
    titulo: Mapped[str] = mapped_column(String(255), nullable=False)
    autor: Mapped[str] = mapped_column(String(150), nullable=False)

    editorial_id: Mapped[int] = mapped_column(
        ForeignKey("editoriales.id"), nullable=False
    )
    genero_id: Mapped[int] = mapped_column(
        ForeignKey("generos.id"), nullable=False
    )

    editorial: Mapped["EditorialModel"] = relationship(
        back_populates="libros"
    )
    genero: Mapped["GeneroModel"] = relationship(
        back_populates="libros"
    )

    precios: Mapped[list["PrecioModel"]] = relationship(
        back_populates="libro"
    )
    stock: Mapped["StockModel | None"] = relationship(
        back_populates="libro",
        uselist=False
    )


class PrecioModel(Base):
    """Representa el precio de un libro en una moneda determinada."""

    __tablename__ = "precios"

    __table_args__ = (
        CheckConstraint("valor > 0", name="ck_precio_positivo"),
    )

    id: Mapped[int] = mapped_column(Integer, primary_key=True)

    libro_id: Mapped[int] = mapped_column(
        ForeignKey("libros.id"), nullable=False
    )
    moneda_id: Mapped[int] = mapped_column(
        ForeignKey("monedas.id"), nullable=False
    )

    valor: Mapped[Decimal] = mapped_column(
        Numeric(15, 2), nullable=False
    )

    libro: Mapped["LibroModel"] = relationship(
        back_populates="precios"
    )
    moneda: Mapped["MonedaModel"] = relationship(
        back_populates="precios"
    )


class StockModel(Base):
    """Representa el stock de un libro mediante una relación uno a uno."""

    __tablename__ = "stock"

    __table_args__ = (
        CheckConstraint("cantidad >= 0", name="ck_stock_no_negativo"),
    )

    id: Mapped[int] = mapped_column(Integer, primary_key=True)

    libro_id: Mapped[int] = mapped_column(
        ForeignKey("libros.id"),
        nullable=False,
        unique=True
    )

    cantidad: Mapped[int] = mapped_column(
        Integer, nullable=False
    )

    libro: Mapped["LibroModel"] = relationship(
        back_populates="stock"
    )


class CotizacionDolarModel(Base):
    """Representa las cotizaciones del dólar por tipo, moneda y fecha."""

    __tablename__ = "cotizaciones_dolar"

    __table_args__ = (
        UniqueConstraint(
            "tipo_cotizacion_id",
            "fecha",
            name="uq_cotizacion_tipo_fecha"
        ),
        CheckConstraint(
            "valor_compra > 0",
            name="ck_cotizacion_compra_positiva"
        ),
        CheckConstraint(
            "valor_venta > 0",
            name="ck_cotizacion_venta_positiva"
        ),
        CheckConstraint(
            "valor_venta >= valor_compra",
            name="ck_cotizacion_valores"
        ),
    )

    id: Mapped[int] = mapped_column(Integer, primary_key=True)

    tipo_cotizacion_id: Mapped[int] = mapped_column(
        ForeignKey("tipos_cotizacion.id"), nullable=False
    )
    moneda_id: Mapped[int] = mapped_column(
        ForeignKey("monedas.id"), nullable=False
    )

    fecha: Mapped[date] = mapped_column(Date, nullable=False)

    valor_compra: Mapped[Decimal] = mapped_column(
        Numeric(15, 2), nullable=False
    )
    valor_venta: Mapped[Decimal] = mapped_column(
        Numeric(15, 2), nullable=False
    )

    tipo_cotizacion: Mapped["TipoCotizacionModel"] = relationship(
        back_populates="cotizaciones"
    )
    moneda: Mapped["MonedaModel"] = relationship(
        back_populates="cotizaciones"
    )
````

### 05 Migración a Sql

````python
# Migración de datos

%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/migrations/migrations.py

"""
Módulo de migración de datos del sistema Book Manager.

Convierte los registros almacenados en archivos CSV
en sentencias SQL INSERT para las tablas del sistema.
"""

import csv
from pathlib import Path
from datetime import date
from decimal import Decimal


def migrar_datos(carpeta_csvs: str, carpeta_sqls: str) -> Path:
    """
    Genera un archivo SQL con los datos de los archivos CSV.

    Procesa las tablas en un orden que respeta las dependencias
    entre claves primarias y foráneas.

    Args:
        carpeta_csvs (str): Ruta de la carpeta que contiene los CSV.
        carpeta_sqls (str): Ruta donde se guardará el archivo SQL.

    Returns:
        Path: Ruta del archivo SQL generado.

    Raises:
        FileNotFoundError: Si falta alguno de los CSV requeridos.
        ValueError: Si un registro contiene datos inválidos.
    """

    ruta_csvs = Path(carpeta_csvs)
    ruta_sqls = Path(carpeta_sqls)

    ruta_sqls.mkdir(parents=True, exist_ok=True)

    archivo_salida = ruta_sqls / "migracion_datos.sql"

    tablas = [
        "generos",
        "editoriales",
        "monedas",
        "tipos_cotizacion",
        "libros",
        "precios",
        "stock",
        "cotizaciones_dolar",
    ]

    columnas_enteras = {
        "id",
        "editorial_id",
        "genero_id",
        "libro_id",
        "moneda_id",
        "tipo_cotizacion_id",
        "cantidad",
    }

    columnas_decimales = {
        "valor",
        "valor_compra",
        "valor_venta",
    }

    sentencias: list[str] = [
        "-- Migración de datos de Book Manager",
        "BEGIN TRANSACTION;",
    ]

    for tabla in tablas:
        archivo_csv = ruta_csvs / f"{tabla}.csv"

        if not archivo_csv.is_file():
            raise FileNotFoundError(
                f"No se encontró el archivo: {archivo_csv}"
            )

        with archivo_csv.open(
            "r", encoding="utf-8-sig", newline=""
        ) as archivo:

            lector = csv.DictReader(archivo)

            if not lector.fieldnames:
                raise ValueError(
                    f"El archivo {archivo_csv.name} no tiene encabezados"
                )

            columnas = lector.fieldnames

            sentencias.append(f"\n-- Tabla: {tabla}")

            for numero_fila, fila in enumerate(lector, start=2):

                valores_sql: list[str] = []

                for columna in columnas:
                    valor = fila[columna]

                    if valor is None or valor == "":
                        valores_sql.append("NULL")

                    elif columna in columnas_enteras:
                        valores_sql.append(str(int(valor)))

                    elif columna in columnas_decimales:
                        valores_sql.append(str(Decimal(valor)))

                    elif columna == "fecha":
                        fecha = date.fromisoformat(valor)
                        valores_sql.append(f"'{fecha.isoformat()}'")

                    else:
                        valor_escapado = valor.replace("'", "''")
                        valores_sql.append(f"'{valor_escapado}'")

                nombres_columnas = ", ".join(columnas)
                valores = ", ".join(valores_sql)

                sentencia = (
                    f"INSERT INTO {tabla} ({nombres_columnas}) "
                    f"VALUES ({valores});"
                )

                sentencias.append(sentencia)

    sentencias.append("\nCOMMIT;")

    archivo_salida.write_text(
        "\n".join(sentencias) + "\n",
        encoding="utf-8",
    )

    return archivo_salida
````

````python
# Modificación del archivo repositorio.py

%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/repositories/repositories.py

"""
Repositorios del sistema Book Manager.

Implementa las operaciones de persistencia mediante SQLAlchemy,
utilizando los modelos ORM y el administrador de conexiones.
"""

from datetime import date
from typing import Generic, TypeVar

from sqlalchemy import select
from sqlalchemy.orm import DeclarativeBase, selectinload

from book_manager.database.connection import ConexionDB
from book_manager.models.models import (
    CotizacionDolarModel,
    StockModel,
)


M = TypeVar("M", bound=DeclarativeBase)


class Repositorio(Generic[M]):
    """
    Repositorio genérico para operaciones CRUD con SQLAlchemy.

    Attributes:
        conexion (ConexionDB): Administrador de conexiones.
        modelo (type[M]): Clase del modelo ORM.
    """

    def __init__(
        self,
        conexion: ConexionDB,
        modelo: type[M],
    ) -> None:
        """Inicializa el repositorio con su conexión y modelo ORM."""
        self._conexion = conexion
        self._modelo = modelo

    def _opciones_relaciones(self):
        """
        Prepara la carga anticipada de relaciones del modelo ORM.

        Permite acceder a las relaciones después de cerrar
        la sesión de SQLAlchemy.
        """
        return [
            selectinload(getattr(self._modelo, relacion.key))
            for relacion in self._modelo.__mapper__.relationships
        ]

    def crear(self, entidad: M) -> M:
        """Inserta un registro en la base de datos."""
        with self._conexion.obtener_sesion() as sesion:
            if entidad.id is not None:
                if sesion.get(self._modelo, entidad.id) is not None:
                    raise ValueError(
                        f"Ya existe un registro con ID {entidad.id}."
                    )

            sesion.add(entidad)
            sesion.flush()
            sesion.refresh(entidad)

        return entidad

    def leer_por_id(self, id: int) -> M | None:
        """Obtiene un registro con sus relaciones cargadas."""
        with self._conexion.obtener_sesion() as sesion:
            consulta = (
                select(self._modelo)
                .options(*self._opciones_relaciones())
                .where(self._modelo.id == id)
            )

            return sesion.scalar(consulta)

    def leer_todos(self) -> list[M]:
        """Obtiene todos los registros con sus relaciones cargadas."""
        with self._conexion.obtener_sesion() as sesion:
            consulta = (
                select(self._modelo)
                .options(*self._opciones_relaciones())
                .order_by(self._modelo.id)
            )

            return list(sesion.scalars(consulta).all())

    def actualizar(self, entidad: M) -> M:
        """Actualiza un registro existente."""
        with self._conexion.obtener_sesion() as sesion:
            if entidad.id is None:
                raise ValueError(
                    "La entidad debe tener un ID para actualizarse."
                )

            if sesion.get(self._modelo, entidad.id) is None:
                raise ValueError(
                    f"No existe un registro con ID {entidad.id}."
                )

            entidad_actualizada = sesion.merge(entidad)
            sesion.flush()

            return entidad_actualizada

    def eliminar(self, id: int) -> bool:
        """Elimina un registro por su identificador."""
        with self._conexion.obtener_sesion() as sesion:
            entidad = sesion.get(self._modelo, id)

            if entidad is None:
                return False

            sesion.delete(entidad)
            sesion.flush()

            return True


class RepositorioStock(Repositorio[StockModel]):
    """
    Repositorio especializado en la gestión de stock.

    Permite consultar y administrar las existencias
    utilizando el identificador del libro.
    """

    def __init__(self, conexion: ConexionDB) -> None:
        """Inicializa el repositorio con el modelo de stock."""
        super().__init__(conexion, StockModel)

    def leer_por_libro(self, libro_id: int) -> StockModel | None:
        """Obtiene el stock correspondiente a un libro."""
        with self._conexion.obtener_sesion() as sesion:
            consulta = (
                select(StockModel)
                .options(*self._opciones_relaciones())
                .where(StockModel.libro_id == libro_id)
            )

            return sesion.scalar(consulta)

    def eliminar_por_libro(self, libro_id: int) -> bool:
        """Elimina el registro de stock correspondiente a un libro."""
        with self._conexion.obtener_sesion() as sesion:
            consulta = select(StockModel).where(
                StockModel.libro_id == libro_id
            )

            stock = sesion.scalar(consulta)

            if stock is None:
                return False

            sesion.delete(stock)
            sesion.flush()

            return True


class RepositorioCotizacionDolar(
    Repositorio[CotizacionDolarModel]
):
    """
    Repositorio especializado en cotizaciones del dólar.

    Permite consultar cotizaciones por tipo y fecha,
    además de recuperar su historial.
    """

    def __init__(self, conexion: ConexionDB) -> None:
        """Inicializa el repositorio con el modelo de cotizaciones."""
        super().__init__(conexion, CotizacionDolarModel)

    def leer_por_tipo_y_fecha(
        self,
        tipo_id: int,
        fecha: date,
    ) -> CotizacionDolarModel | None:
        """Obtiene una cotización por tipo y fecha."""
        with self._conexion.obtener_sesion() as sesion:
            consulta = (
                select(CotizacionDolarModel)
                .options(*self._opciones_relaciones())
                .where(
                    CotizacionDolarModel.tipo_cotizacion_id == tipo_id,
                    CotizacionDolarModel.fecha == fecha,
                )
            )

            return sesion.scalar(consulta)

    def leer_historico_por_tipo(
        self,
        tipo_id: int,
    ) -> list[CotizacionDolarModel]:
        """Obtiene el historial de cotizaciones de un tipo."""
        with self._conexion.obtener_sesion() as sesion:
            consulta = (
                select(CotizacionDolarModel)
                .options(*self._opciones_relaciones())
                .where(
                    CotizacionDolarModel.tipo_cotizacion_id == tipo_id
                )
                .order_by(CotizacionDolarModel.fecha)
            )

            return list(sesion.scalars(consulta).all())

    def eliminar_por_tipo_y_fecha(
        self,
        tipo_id: int,
        fecha: date,
    ) -> bool:
        """Elimina una cotización identificada por tipo y fecha."""
        with self._conexion.obtener_sesion() as sesion:
            consulta = select(CotizacionDolarModel).where(
                CotizacionDolarModel.tipo_cotizacion_id == tipo_id,
                CotizacionDolarModel.fecha == fecha,
            )

            cotizacion = sesion.scalar(consulta)

            if cotizacion is None:
                return False

            sesion.delete(cotizacion)
            sesion.flush()

            return True
````

#### services.py

````python
# Modificacion del archivo services
%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/services/services.py

"""
Servicios del sistema Book Manager.

Contiene la lógica de negocio y las validaciones necesarias
para operar con los modelos ORM mediante los repositorios.
"""

import os
from datetime import date, datetime
from decimal import Decimal
from pathlib import Path
from typing import Generic, TypeVar
from zoneinfo import ZoneInfo

import requests
from dotenv import load_dotenv
from sqlalchemy.orm import DeclarativeBase

from book_manager.models.models import (
    LibroModel,
    PrecioModel,
    StockModel,
    CotizacionDolarModel,
)

from book_manager.repositories.repositories import (
    Repositorio,
    RepositorioStock,
    RepositorioCotizacionDolar,
)


M = TypeVar("M", bound=DeclarativeBase)


class ServicioBase(Generic[M]):
    """Servicio genérico para operaciones CRUD."""

    def __init__(self, repositorio: Repositorio[M]) -> None:
        self._repositorio = repositorio

    def crear(self, entidad: M) -> M:
        return self._repositorio.crear(entidad)

    def obtener_por_id(self, id: int) -> M | None:
        return self._repositorio.leer_por_id(id)

    def obtener_todos(self) -> list[M]:
        return self._repositorio.leer_todos()

    def actualizar(self, entidad: M) -> M:
        return self._repositorio.actualizar(entidad)

    def eliminar(self, id: int) -> bool:
        return self._repositorio.eliminar(id)


class ServicioLibro(ServicioBase[LibroModel]):
    """Servicio de gestión de libros."""

    def crear(self, libro: LibroModel) -> LibroModel:

        if not libro.isbn or not libro.isbn.strip():
            raise ValueError("El ISBN no puede estar vacío.")

        if not libro.titulo or not libro.titulo.strip():
            raise ValueError("El título no puede estar vacío.")

        if not libro.autor or not libro.autor.strip():
            raise ValueError("El autor no puede estar vacío.")

        return self._repositorio.crear(libro)


class ServicioPrecio(ServicioBase[PrecioModel]):
    """Servicio de gestión de precios."""

    def crear(self, precio: PrecioModel) -> PrecioModel:

        if precio.valor is None or precio.valor <= 0:
            raise ValueError("El precio debe ser mayor que cero.")

        return self._repositorio.crear(precio)


class ServicioStock:
    """Servicio de gestión de existencias."""

    def __init__(self, repositorio: RepositorioStock) -> None:
        self._repositorio = repositorio

    def listar(self) -> list[StockModel]:
        """Obtiene todos los registros de stock."""
        return self._repositorio.leer_todos()

    def crear(self, stock: StockModel) -> StockModel:

        if stock.cantidad is None or stock.cantidad < 0:
            raise ValueError("El stock no puede ser negativo.")

        return self._repositorio.crear(stock)

    def obtener_por_libro(self, libro_id: int) -> StockModel | None:
        return self._repositorio.leer_por_libro(libro_id)

    def actualizar(self, stock: StockModel) -> StockModel:

        if stock.cantidad is None or stock.cantidad < 0:
            raise ValueError("El stock no puede ser negativo.")

        return self._repositorio.actualizar(stock)

    def eliminar(self, libro_id: int) -> bool:
        return self._repositorio.eliminar_por_libro(libro_id)


class ServicioCotizacionDolar:
    """Servicio de gestión de cotizaciones del dólar."""

    def __init__(
        self,
        repositorio: RepositorioCotizacionDolar,
    ) -> None:
        self._repositorio = repositorio

    def _validar(
        self,
        cotizacion: CotizacionDolarModel,
    ) -> None:
        """Valida los valores de compra y venta."""

        if cotizacion.valor_compra is None or cotizacion.valor_compra <= 0:
            raise ValueError(
                "El valor de compra debe ser mayor que cero."
            )

        if cotizacion.valor_venta is None or cotizacion.valor_venta <= 0:
            raise ValueError(
                "El valor de venta debe ser mayor que cero."
            )

        if cotizacion.valor_venta < cotizacion.valor_compra:
            raise ValueError(
                "El valor de venta no puede ser menor que el valor de compra."
            )

    def crear(
        self,
        cotizacion: CotizacionDolarModel,
    ) -> CotizacionDolarModel:
        """Valida y registra una cotización."""

        self._validar(cotizacion)
        return self._repositorio.crear(cotizacion)

    def obtener_por_tipo_y_fecha(
        self,
        tipo_id: int,
        fecha: date,
    ) -> CotizacionDolarModel | None:

        return self._repositorio.leer_por_tipo_y_fecha(
            tipo_id,
            fecha,
        )

    def obtener_historico(
        self,
        tipo_id: int,
    ) -> list[CotizacionDolarModel]:

        return self._repositorio.leer_historico_por_tipo(tipo_id)

    def actualizar(
        self,
        cotizacion: CotizacionDolarModel,
    ) -> CotizacionDolarModel:

        self._validar(cotizacion)
        return self._repositorio.actualizar(cotizacion)

    def eliminar(
        self,
        tipo_id: int,
        fecha: date,
    ) -> bool:

        return self._repositorio.eliminar_por_tipo_y_fecha(
            tipo_id,
            fecha,
        )

    def obtener_cotizaciones(self) -> dict[str, int]:
        """
        Descarga cotizaciones desde DolarAPI.

        Crea nuevas cotizaciones y actualiza las existentes
        según el tipo de dólar y la fecha.
        """

        # 1. Leer configuración del archivo .env

        ruta_env = Path(__file__).resolve().parents[3] / ".env"

        load_dotenv(ruta_env)

        api_url = os.getenv("API_URL")

        if not api_url:
            raise ValueError(
                "No se encontró API_URL en el archivo .env."
            )

        # 2. Consultar la API

        respuesta = requests.get(api_url, timeout=15)
        respuesta.raise_for_status()

        datos = respuesta.json()

        if not isinstance(datos, list):
            raise ValueError(
                "La API no devolvió una lista de cotizaciones."
            )

        # 3. Relacionar los tipos de dólar con nuestra BD

        tipos = {
            "oficial": 1,
            "blue": 2,
            "bolsa": 3,
            "contadoconliqui": 4,
            "mayorista": 5,
            "cripto": 6,
            "tarjeta": 8,
        }

        moneda_usd_id = 2

        # 4. Inicializar contadores

        creadas = 0
        actualizadas = 0
        omitidas = 0

        # 5. Procesar cada cotización

        for dato in datos:

            tipo_id = tipos.get(dato.get("casa"))

            if tipo_id is None or dato.get("moneda") != "USD":
                omitidas += 1
                continue

            # Convertir la fecha UTC a fecha argentina

            fecha_actualizacion = datetime.fromisoformat(
                dato["fechaActualizacion"].replace("Z", "+00:00")
            )

            fecha_cotizacion = fecha_actualizacion.astimezone(
                ZoneInfo("America/Argentina/Buenos_Aires")
            ).date()

            # Convertir importes a Decimal

            compra = Decimal(str(dato["compra"]))
            venta = Decimal(str(dato["venta"]))

            # Buscar cotización existente

            existente = self.obtener_por_tipo_y_fecha(
                tipo_id,
                fecha_cotizacion,
            )

            if existente is None:

                # 6. Crear una nueva cotización

                cotizacion = CotizacionDolarModel(
                    tipo_cotizacion_id=tipo_id,
                    moneda_id=moneda_usd_id,
                    fecha=fecha_cotizacion,
                    valor_compra=compra,
                    valor_venta=venta,
                )

                self.crear(cotizacion)
                creadas += 1

            else:

                # 7. Actualizar una cotización existente

                existente.moneda_id = moneda_usd_id
                existente.valor_compra = compra
                existente.valor_venta = venta

                self.actualizar(existente)
                actualizadas += 1

        # 8. Devolver resumen

        return {
            "creadas": creadas,
            "actualizadas": actualizadas,
            "omitidas": omitidas,
        }
````

````python
# Modifiación main
%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/main.py

"""
Punto de entrada del sistema Book Manager.

Inicializa SQLite, los repositorios, los servicios
y la interfaz de consola.
"""

from book_manager.database.connection import ConexionDB

from book_manager.models.models import (
    Base,
    GeneroModel,
    EditorialModel,
    MonedaModel,
    TipoCotizacionModel,
    LibroModel,
    PrecioModel,
)

from book_manager.preload_data.preload_data import cargar_datos

from book_manager.repositories.repositories import (
    Repositorio,
    RepositorioStock,
    RepositorioCotizacionDolar,
)

from book_manager.services.services import (
    ServicioBase,
    ServicioCotizacionDolar,
    ServicioLibro,
    ServicioPrecio,
    ServicioStock,
)

from book_manager.ui.console import Consola
from pathlib import Path

def main(import_default_data: bool = False) -> None:
    """
    Inicializa y ejecuta Book Manager.

    Args:
        import_default_data: Si es True, importa los datos
            iniciales desde los archivos CSV cuando la
            base de datos está vacía.
    """

    # -------------------------------------------
    # 1. Inicialización de base de datos
    # -------------------------------------------

    conexion = ConexionDB()

    Base.metadata.create_all(conexion.engine)

    print(f"Base de datos utilizada: {conexion.engine.dialect.name}")

    # -------------------------------------------
    # 2. Carga opcional de datos iniciales
    # -------------------------------------------

    if import_default_data:
        cargar_datos(conexion)

    # -------------------------------------------
    # 3. Inicialización de repositorios
    # -------------------------------------------

    repositorios = {
        "generos": Repositorio(conexion, GeneroModel),
        "editoriales": Repositorio(conexion, EditorialModel),
        "monedas": Repositorio(conexion, MonedaModel),
        "tipos_cotizacion": Repositorio(
            conexion, TipoCotizacionModel
        ),
        "libros": Repositorio(conexion, LibroModel),
        "precios": Repositorio(conexion, PrecioModel),
        "stock": RepositorioStock(conexion),
        "cotizaciones": RepositorioCotizacionDolar(conexion),
    }

    # -------------------------------------------
    # 4. Inicialización de servicios
    # -------------------------------------------

    servicio_generos = ServicioBase(
        repositorios["generos"]
    )

    servicio_editoriales = ServicioBase(
        repositorios["editoriales"]
    )

    servicio_monedas = ServicioBase(
        repositorios["monedas"]
    )

    servicio_tipos_cotizacion = ServicioBase(
        repositorios["tipos_cotizacion"]
    )

    servicio_libros = ServicioLibro(
        repositorios["libros"]
    )

    servicio_precios = ServicioPrecio(
        repositorios["precios"]
    )

    servicio_stock = ServicioStock(
        repositorios["stock"]
    )

    servicio_cotizaciones = ServicioCotizacionDolar(
        repositorios["cotizaciones"]
    )

    # -------------------------------------------
    # 5. Inicialización de la consola
    # -------------------------------------------

    consola = Consola(
        servicio_libros=servicio_libros,
        servicio_editoriales=servicio_editoriales,
        servicio_generos=servicio_generos,
        servicio_precios=servicio_precios,
        servicio_monedas=servicio_monedas,
        servicio_stock=servicio_stock,
        servicio_cotizaciones=servicio_cotizaciones,
        servicio_tipos_cotizacion=servicio_tipos_cotizacion,
    )

    # -------------------------------------------
    # 6. Ejecución
    # -------------------------------------------

    consola.ejecutar()


if __name__ == "__main__":
    main()
````

````python
# Modifiación del archivo preload

%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/preload_data/preload_data.py

"""
Carga inicial de datos del sistema Book Manager.

Utiliza la migración SQL generada a partir de los CSV
para poblar una base de datos SQLite mediante SQLAlchemy.
"""

from pathlib import Path

from sqlalchemy import text

from book_manager.database.connection import ConexionDB
from book_manager.migrations.migrations import migrar_datos
from book_manager.models.models import Base, GeneroModel


def cargar_datos(
    conexion: ConexionDB,
    ruta_base: str | Path | None = None,
) -> None:
    """
    Carga los registros iniciales en SQLite si la base está vacía.

    Args:
        conexion (ConexionDB): Administrador de conexiones SQLite.
        ruta_base (str | Path | None): Carpeta de archivos CSV.

    Raises:
        FileNotFoundError: Si no existen los archivos necesarios.
        ValueError: Si la base contiene una carga parcial.
    """
    ruta_migrations = Path(__file__).resolve().parent.parent / "migrations"

    ruta_csv = (
        Path(ruta_base)
        if ruta_base is not None
        else ruta_migrations / "csv"
    )

    ruta_sql = ruta_migrations / "sql"

    Base.metadata.create_all(conexion.engine)

    # Verificar el estado de las ocho tablas.
    with conexion.obtener_sesion() as sesion:
        tablas = list(Base.metadata.sorted_tables)

        cantidades = {
            tabla.name: sesion.execute(
                text(f'SELECT COUNT(*) FROM "{tabla.name}"')
            ).scalar_one()
            for tabla in tablas
        }

    if all(cantidad > 0 for cantidad in cantidades.values()):
        print("La base ya contiene datos. No se repite la carga.")
        return

    if any(cantidad > 0 for cantidad in cantidades.values()):
        raise ValueError(
            "La base contiene datos parciales. "
            "Se cancela la carga inicial para evitar duplicados."
        )

    # Generar el archivo SQL utilizando la migración existente.
    archivo_sql = migrar_datos(
        str(ruta_csv),
        str(ruta_sql),
    )

    contenido_sql = archivo_sql.read_text(encoding="utf-8")

    # Ejecutar el script completo dentro de una transacción SQLite.
    # La migración existente contiene instrucciones de transacción.
    conexion_sqlite = conexion.engine.raw_connection()

    try:
        conexion_sqlite.executescript(contenido_sql)
        conexion_sqlite.commit()
    except Exception:
        conexion_sqlite.rollback()
        raise
    finally:
        conexion_sqlite.close()

    print("Datos iniciales cargados correctamente en SQLite.")
````

````python
# Modificación: menu Precios
%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/ui/menus/menu_precios.py

"""
Interfaz de consola para la gestión de precios con SQLAlchemy.
"""

from decimal import Decimal, InvalidOperation

from sqlalchemy.exc import IntegrityError

from book_manager.models.models import PrecioModel
from book_manager.services.services import ServicioBase, ServicioPrecio


class MenuPrecios:
    """Interfaz de consola para la gestión de precios."""

    def __init__(
        self,
        servicio_precios: ServicioPrecio,
        servicio_libros: ServicioBase,
        servicio_monedas: ServicioBase,
    ) -> None:
        """Inicializa los servicios necesarios para gestionar precios."""

        self.servicio_precios = servicio_precios
        self.servicio_libros = servicio_libros
        self.servicio_monedas = servicio_monedas

    def ejecutar(self) -> None:
        """Ejecuta el menú de gestión de precios."""

        while True:
            self.mostrar_menu()

            opcion = input("Seleccione una opción: ").strip()

            if opcion == "1":
                self.listar_precios()
            elif opcion == "2":
                self.buscar_precio()
            elif opcion == "3":
                self.crear_precio()
            elif opcion == "4":
                self.modificar_precio()
            elif opcion == "5":
                self.eliminar_precio()
            elif opcion == "0":
                break
            else:
                print("\nOpción inválida. Intente nuevamente.")

    @staticmethod
    def mostrar_menu() -> None:
        """Muestra las opciones de gestión de precios."""

        print("\n" + "=" * 40)
        print("         GESTIÓN DE PRECIOS")
        print("=" * 40)
        print("1. Listar precios")
        print("2. Buscar precio")
        print("3. Crear precio")
        print("4. Modificar precio")
        print("5. Eliminar precio")
        print("0. Volver")
        print("=" * 40)

    @staticmethod
    def mostrar_precio(precio: PrecioModel) -> None:
        """Muestra un precio junto con sus relaciones."""

        print(
            f"ID: {precio.id} | "
            f"Libro: {precio.libro.titulo} | "
            f"Moneda: {precio.moneda.codigo} | "
            f"Valor: {precio.moneda.simbolo}{precio.valor:.2f}"
        )

    @staticmethod
    def solicitar_valor(mensaje: str) -> Decimal:
        """Solicita un importe positivo con precisión decimal."""

        texto = input(mensaje).strip().replace(",", ".")

        try:
            valor = Decimal(texto)
        except InvalidOperation as error:
            raise ValueError(
                "El valor debe ser un número válido."
            ) from error

        if not valor.is_finite() or valor <= 0:
            raise ValueError(
                "El precio debe ser un número finito mayor que cero."
            )

        if valor.as_tuple().exponent < -2:
            raise ValueError(
                "El precio puede tener como máximo dos decimales."
            )

        if valor > Decimal("9999999999999.99"):
            raise ValueError(
                "El precio supera el máximo permitido."
            )

        return valor

    def listar_precios(self) -> None:
        """Lista todos los precios."""

        precios = self.servicio_precios.obtener_todos()

        if not precios:
            print("\nNo hay precios cargados.")
            return

        print("\n--- LISTADO DE PRECIOS ---")

        for precio in precios:
            self.mostrar_precio(precio)

    def buscar_precio(self) -> None:
        """Busca un precio por su ID."""

        try:
            precio_id = int(input("Ingrese el ID del precio: "))

            precio = self.servicio_precios.obtener_por_id(precio_id)

            if precio is None:
                print("\nPrecio no encontrado.")
                return

            print("\n--- PRECIO ENCONTRADO ---")

            self.mostrar_precio(precio)

        except ValueError:
            print("\nEl ID debe ser un número entero.")

    def crear_precio(self) -> None:
        """Crea un nuevo precio mediante claves foráneas ORM."""

        try:
            print("\n--- CREAR PRECIO ---")

            precio_id = int(input("ID: "))
            libro_id = int(input("ID del libro: "))
            moneda_id = int(input("ID de la moneda: "))

            valor = self.solicitar_valor("Valor: ")

            libro = self.servicio_libros.obtener_por_id(libro_id)
            moneda = self.servicio_monedas.obtener_por_id(moneda_id)

            if libro is None:
                print("\nEl libro indicado no existe.")
                return

            if moneda is None:
                print("\nLa moneda indicada no existe.")
                return

            precio = PrecioModel(
                id=precio_id,
                libro_id=libro_id,
                moneda_id=moneda_id,
                valor=valor,
            )

            self.servicio_precios.crear(precio)

            print("\nPrecio creado correctamente.")

        except (ValueError, IntegrityError) as error:
            print(f"\nError al crear el precio: {error}")

    def modificar_precio(self) -> None:
        """Modifica el importe de un precio existente."""

        try:
            print("\n--- MODIFICAR PRECIO ---")

            precio_id = int(input("ID del precio a modificar: "))

            precio_actual = self.servicio_precios.obtener_por_id(
                precio_id
            )

            if precio_actual is None:
                print("\nPrecio no encontrado.")
                return

            valor_ingresado = input(
                f"Valor [{precio_actual.valor}]: "
            ).strip()

            if valor_ingresado:
                # Reutilizamos la validación del importe.
                # Se utiliza el texto ya ingresado, sin pedirlo dos veces.
                texto = valor_ingresado.replace(",", ".")

                try:
                    valor = Decimal(texto)
                except InvalidOperation as error:
                    raise ValueError(
                        "El valor debe ser un número válido."
                    ) from error

                if not valor.is_finite() or valor <= 0:
                    raise ValueError(
                        "El precio debe ser mayor que cero."
                    )

                if valor.as_tuple().exponent < -2:
                    raise ValueError(
                        "El precio puede tener como máximo dos decimales."
                    )

                if valor > Decimal("9999999999999.99"):
                    raise ValueError(
                        "El precio supera el máximo permitido."
                    )
            else:
                valor = precio_actual.valor

            precio_modificado = PrecioModel(
                id=precio_actual.id,
                libro_id=precio_actual.libro_id,
                moneda_id=precio_actual.moneda_id,
                valor=valor,
            )

            self.servicio_precios.actualizar(precio_modificado)

            print("\nPrecio modificado correctamente.")

        except (ValueError, IntegrityError) as error:
            print(f"\nError al modificar el precio: {error}")

    def eliminar_precio(self) -> None:
        """Elimina un precio por su ID."""

        try:
            print("\n--- ELIMINAR PRECIO ---")

            precio_id = int(input("ID del precio a eliminar: "))

            precio = self.servicio_precios.obtener_por_id(precio_id)

            if precio is None:
                print("\nPrecio no encontrado.")
                return

            print(
                f"\nPrecio encontrado: "
                f"{precio.libro.titulo} - "
                f"{precio.moneda.simbolo}{precio.valor:.2f}"
            )

            confirmacion = input(
                "¿Confirma la eliminación? (s/n): "
            ).strip().lower()

            if confirmacion != "s":
                print("\nEliminación cancelada.")
                return

            self.servicio_precios.eliminar(precio_id)

            print("\nPrecio eliminado correctamente.")

        except (ValueError, IntegrityError) as error:
            print(f"\nError al eliminar el precio: {error}")
````

````python
# Modifiación: menu stock

%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/ui/menus/menu_stock.py

"""
Interfaz de consola para la gestión de stock con SQLAlchemy.
"""

from sqlalchemy.exc import IntegrityError

from book_manager.models.models import StockModel
from book_manager.services.services import ServicioBase, ServicioStock


class MenuStock:
    """Interfaz de consola para la gestión de stock."""

    def __init__(
        self,
        servicio_stock: ServicioStock,
        servicio_libros: ServicioBase,
    ) -> None:
        """Inicializa los servicios necesarios para gestionar stock."""
        self.servicio_stock = servicio_stock
        self.servicio_libros = servicio_libros

    def ejecutar(self) -> None:
        """Ejecuta el menú de gestión de stock."""
        while True:
            self.mostrar_menu()
            opcion = input("Seleccione una opción: ").strip()

            if opcion == "1":
                self.listar_stock()
            elif opcion == "2":
                self.buscar_stock()
            elif opcion == "3":
                self.crear_stock()
            elif opcion == "4":
                self.modificar_stock()
            elif opcion == "5":
                self.eliminar_stock()
            elif opcion == "0":
                break
            else:
                print("\nOpción inválida. Intente nuevamente.")

    @staticmethod
    def mostrar_menu() -> None:
        """Muestra las opciones de gestión de stock."""
        print("\n" + "=" * 40)
        print("          GESTIÓN DE STOCK")
        print("=" * 40)
        print("1. Listar stock")
        print("2. Buscar stock por libro")
        print("3. Crear stock")
        print("4. Modificar stock")
        print("5. Eliminar stock")
        print("0. Volver")
        print("=" * 40)

    @staticmethod
    def mostrar_stock(stock: StockModel) -> None:
        """Muestra los datos del stock y del libro relacionado."""
        print(f"Stock ID: {stock.id}")
        print(f"Libro ID: {stock.libro_id}")
        print(f"Título: {stock.libro.titulo}")
        print(f"Cantidad: {stock.cantidad}")

    @staticmethod
    def validar_cantidad(cantidad: int) -> None:
        """Verifica que la cantidad de stock no sea negativa."""
        if cantidad < 0:
            raise ValueError(
                "La cantidad de stock no puede ser negativa."
            )

    def listar_stock(self) -> None:
        """Lista todos los registros de stock."""

        print("\n--- LISTADO DE STOCK ---")

        stocks = self.servicio_stock.listar()

        if not stocks:
            print("\nNo hay registros de stock.")
            return

        for stock in stocks:
            print("\n" + "-" * 40)
            self.mostrar_stock(stock)

        print("\n" + "-" * 40)
        print(f"Total de registros: {len(stocks)}")

    def buscar_stock(self) -> None:
        """Busca el stock correspondiente a un libro."""
        try:
            libro_id = int(input("Ingrese el ID del libro: "))
            stock = self.servicio_stock.obtener_por_libro(libro_id)

            if stock is None:
                print("\nNo se encontró stock para ese libro.")
                return

            print("\n--- STOCK ENCONTRADO ---")
            self.mostrar_stock(stock)

        except ValueError as error:
            print(f"\nError: {error}")

    def crear_stock(self) -> None:
        """Crea un registro de stock para un libro."""
        try:
            print("\n--- CREAR STOCK ---")

            stock_id = int(input("ID del stock: "))
            libro_id = int(input("ID del libro: "))
            cantidad = int(input("Cantidad: "))

            self.validar_cantidad(cantidad)

            libro = self.servicio_libros.obtener_por_id(libro_id)

            if libro is None:
                print("\nEl libro indicado no existe.")
                return

            if self.servicio_stock.obtener_por_libro(libro_id) is not None:
                print("\nEl libro ya tiene un registro de stock.")
                return

            stock = StockModel(
                id=stock_id,
                libro_id=libro_id,
                cantidad=cantidad,
            )

            self.servicio_stock.crear(stock)
            print("\nStock creado correctamente.")

        except (ValueError, IntegrityError) as error:
            print(f"\nError al crear el stock: {error}")

    def modificar_stock(self) -> None:
        """Modifica la cantidad de stock de un libro."""
        try:
            print("\n--- MODIFICAR STOCK ---")

            libro_id = int(input("ID del libro: "))

            stock_actual = self.servicio_stock.obtener_por_libro(
                libro_id
            )

            if stock_actual is None:
                print("\nNo se encontró stock para ese libro.")
                return

            print(f"Libro: {stock_actual.libro.titulo}")
            print(f"Cantidad actual: {stock_actual.cantidad}")

            cantidad = int(input("Nueva cantidad: "))
            self.validar_cantidad(cantidad)

            stock_modificado = StockModel(
                id=stock_actual.id,
                libro_id=stock_actual.libro_id,
                cantidad=cantidad,
            )

            self.servicio_stock.actualizar(stock_modificado)
            print("\nStock modificado correctamente.")

        except (ValueError, IntegrityError) as error:
            print(f"\nError al modificar el stock: {error}")

    def eliminar_stock(self) -> None:
        """Elimina el stock correspondiente a un libro."""
        try:
            print("\n--- ELIMINAR STOCK ---")

            libro_id = int(input("ID del libro: "))

            stock = self.servicio_stock.obtener_por_libro(libro_id)

            if stock is None:
                print("\nNo se encontró stock para ese libro.")
                return

            print(f"\nLibro: {stock.libro.titulo}")
            print(f"Cantidad actual: {stock.cantidad}")

            confirmacion = input(
                "¿Confirma la eliminación? (s/n): "
            ).strip().lower()

            if confirmacion != "s":
                print("\nEliminación cancelada.")
                return

            eliminado = self.servicio_stock.eliminar(libro_id)

            if eliminado:
                print("\nStock eliminado correctamente.")
            else:
                print("\nNo se encontró stock para eliminar.")

        except (ValueError, IntegrityError) as error:
            print(f"\nError al eliminar el stock: {error}")
````

#### menu_cotizaciones.py

````python
# Modifiación: menu Cotizaciones

%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/ui/menus/menu_cotizaciones.py
"""
Menú de gestión de cotizaciones del dólar.

Permite consultar, crear, modificar y eliminar cotizaciones
utilizando los servicios y modelos SQLAlchemy.
"""

from datetime import date
from decimal import Decimal, InvalidOperation

from sqlalchemy.exc import IntegrityError

from book_manager.models.models import CotizacionDolarModel


class MenuCotizaciones:
    """Interfaz de consola para gestionar cotizaciones."""

    def __init__(
        self,
        servicio_cotizaciones,
        servicio_tipos_cotizacion,
        servicio_monedas,
    ) -> None:
        """Inicializa el menú con sus servicios."""
        self.servicio = servicio_cotizaciones
        self.servicio_tipos_cotizacion = servicio_tipos_cotizacion
        self.servicio_monedas = servicio_monedas

    @staticmethod
    def leer_id(mensaje: str) -> int:
        """Solicita un identificador positivo."""
        try:
            identificador = int(input(mensaje))
        except ValueError as error:
            raise ValueError("El ID debe ser numérico.") from error

        if identificador <= 0:
            raise ValueError("El ID debe ser mayor que cero.")

        return identificador

    @staticmethod
    def leer_fecha(mensaje: str) -> date:
        """Solicita una fecha en formato AAAA-MM-DD."""
        try:
            return date.fromisoformat(input(mensaje).strip())
        except ValueError as error:
            raise ValueError(
                "La fecha debe tener formato AAAA-MM-DD."
            ) from error

    @staticmethod
    def validar_decimal(valor: str) -> Decimal:
        """Valida un importe positivo con hasta dos decimales."""
        try:
            numero = Decimal(valor.strip().replace(",", "."))
        except (InvalidOperation, AttributeError) as error:
            raise ValueError(
                "El importe debe ser numérico."
            ) from error

        if not numero.is_finite() or numero <= 0:
            raise ValueError(
                "El importe debe ser mayor que cero."
            )

        if numero.as_tuple().exponent < -2:
            raise ValueError(
                "El importe puede tener como máximo dos decimales."
            )

        if numero > Decimal("9999999999999.99"):
            raise ValueError(
                "El importe supera el máximo permitido."
            )

        return numero

    @staticmethod
    def validar_valores(
        valor_compra: Decimal,
        valor_venta: Decimal,
    ) -> None:
        """Valida la coherencia de los valores de cotización."""
        if valor_compra <= 0 or valor_venta <= 0:
            raise ValueError(
                "Los valores deben ser mayores que cero."
            )

        if valor_compra > valor_venta:
            raise ValueError(
                "El valor de compra no puede superar al de venta."
            )

    def validar_referencias(
        self,
        tipo_id: int,
        moneda_id: int,
    ) -> None:
        """Comprueba que el tipo de cotización y la moneda existan."""
        tipo = self.servicio_tipos_cotizacion.obtener_por_id(
            tipo_id
        )

        if tipo is None:
            raise ValueError(
                "El tipo de cotización no existe."
            )

        moneda = self.servicio_monedas.obtener_por_id(
            moneda_id
        )

        if moneda is None:
            raise ValueError(
                "La moneda no existe."
            )

    def ejecutar(self) -> None:
        """Ejecuta el menú de gestión de cotizaciones."""
        opciones = {
            "1": self.listar,
            "2": self.buscar,
            "3": self.crear,
            "4": self.modificar,
            "5": self.eliminar,
        }

        while True:
            print("\n--- GESTIÓN DE COTIZACIONES ---")
            print("1. Listar historial por tipo")
            print("2. Buscar por tipo y fecha")
            print("3. Crear cotización")
            print("4. Modificar cotización")
            print("5. Eliminar cotización")
            print("0. Volver")

            opcion = input(
                "Seleccione una opción: "
            ).strip()

            if opcion == "0":
                return

            accion = opciones.get(opcion)

            if accion is None:
                print("\nOpción inválida.")
                continue

            try:
                accion()

            except ValueError as error:
                print(f"\nDatos inválidos: {error}")

            except IntegrityError:
                print(
                    "\nNo se pudo completar la operación. "
                    "Verificá que no exista una cotización "
                    "duplicada y que las referencias sean válidas."
                )

    def listar(self) -> None:
        """Lista el historial de cotizaciones de un tipo."""
        tipos = self.servicio_tipos_cotizacion.obtener_todos()

        if not tipos:
            print("\nNo hay tipos de cotización registrados.")
            return

        print("\n--- TIPOS DE COTIZACIÓN ---")

        for tipo in tipos:
            print(f"{tipo.id}. {tipo.nombre}")

        tipo_id = self.leer_id(
            "\nID del tipo de cotización: "
        )

        tipo = self.servicio_tipos_cotizacion.obtener_por_id(
            tipo_id
        )

        if tipo is None:
            print("\nTipo de cotización inexistente.")
            return

        cotizaciones = self.servicio.obtener_historico(
            tipo_id
        )

        if not cotizaciones:
            print("\nNo hay cotizaciones registradas.")
            return

        print(f"\n--- HISTORIAL: {tipo.nombre} ---")

        for cotizacion in cotizaciones:
            self.mostrar_cotizacion(cotizacion)

    def buscar(self) -> None:
        """Busca una cotización por tipo y fecha."""
        tipo_id = self.leer_id(
            "ID del tipo de cotización: "
        )

        fecha = self.leer_fecha(
            "Fecha (AAAA-MM-DD): "
        )

        cotizacion = self.servicio.obtener_por_tipo_y_fecha(
            tipo_id,
            fecha,
        )

        if cotizacion is None:
            print("\nCotización no encontrada.")
            return

        self.mostrar_cotizacion(cotizacion)

    def crear(self) -> None:
        """Registra una nueva cotización."""
        print("\n--- NUEVA COTIZACIÓN ---")

        tipo_id = self.leer_id(
            "ID del tipo de cotización: "
        )

        moneda_id = self.leer_id(
            "ID de la moneda: "
        )

        self.validar_referencias(
            tipo_id,
            moneda_id,
        )

        fecha = self.leer_fecha(
            "Fecha (AAAA-MM-DD): "
        )

        valor_compra = self.validar_decimal(
            input("Valor compra: ")
        )

        valor_venta = self.validar_decimal(
            input("Valor venta: ")
        )

        self.validar_valores(
            valor_compra,
            valor_venta,
        )

        existente = self.servicio.obtener_por_tipo_y_fecha(
            tipo_id,
            fecha,
        )

        if existente is not None:
            print(
                "\nYa existe una cotización para "
                "ese tipo y fecha."
            )
            return

        cotizacion = CotizacionDolarModel(
            tipo_cotizacion_id=tipo_id,
            moneda_id=moneda_id,
            fecha=fecha,
            valor_compra=valor_compra,
            valor_venta=valor_venta,
        )

        self.servicio.crear(cotizacion)

        print("\nCotización creada correctamente.")

    def modificar(self) -> None:
        """Modifica una cotización existente."""
        print("\n--- MODIFICAR COTIZACIÓN ---")

        tipo_id = self.leer_id(
            "ID del tipo de cotización: "
        )

        fecha = self.leer_fecha(
            "Fecha (AAAA-MM-DD): "
        )

        cotizacion = self.servicio.obtener_por_tipo_y_fecha(
            tipo_id,
            fecha,
        )

        if cotizacion is None:
            print("\nCotización no encontrada.")
            return

        self.mostrar_cotizacion(cotizacion)

        moneda_id = self.leer_id(
            "\nNuevo ID de moneda: "
        )

        self.validar_referencias(
            tipo_id,
            moneda_id,
        )

        valor_compra = self.validar_decimal(
            input("Nuevo valor compra: ")
        )

        valor_venta = self.validar_decimal(
            input("Nuevo valor venta: ")
        )

        self.validar_valores(
            valor_compra,
            valor_venta,
        )

        cotizacion.moneda_id = moneda_id
        cotizacion.valor_compra = valor_compra
        cotizacion.valor_venta = valor_venta

        self.servicio.actualizar(cotizacion)

        print("\nCotización modificada correctamente.")

    def eliminar(self) -> None:
        """Elimina una cotización por tipo y fecha."""
        print("\n--- ELIMINAR COTIZACIÓN ---")

        tipo_id = self.leer_id(
            "ID del tipo de cotización: "
        )

        fecha = self.leer_fecha(
            "Fecha (AAAA-MM-DD): "
        )

        cotizacion = self.servicio.obtener_por_tipo_y_fecha(
            tipo_id,
            fecha,
        )

        if cotizacion is None:
            print("\nCotización no encontrada.")
            return

        self.mostrar_cotizacion(cotizacion)

        confirmacion = input(
            "\n¿Confirmar eliminación? (s/n): "
        ).strip().lower()

        if confirmacion != "s":
            print("\nOperación cancelada.")
            return

        eliminado = self.servicio.eliminar(
            tipo_id,
            fecha,
        )

        if eliminado:
            print("\nCotización eliminada correctamente.")
        else:
            print("\nNo se pudo eliminar la cotización.")

    @staticmethod
    def mostrar_cotizacion(
        cotizacion: CotizacionDolarModel,
    ) -> None:
        """Muestra los datos de una cotización."""
        print("\n--- COTIZACIÓN ---")
        print(f"ID: {cotizacion.id}")
        print(
            f"Tipo: {cotizacion.tipo_cotizacion.nombre}"
        )
        print(
            f"Moneda: {cotizacion.moneda.nombre}"
        )
        print(f"Fecha: {cotizacion.fecha}")
        print(
            f"Valor compra: {cotizacion.valor_compra:.2f}"
        )
        print(
            f"Valor venta: {cotizacion.valor_venta:.2f}"
        )
````

````python
# Modifiación: menu Catalogo

%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/ui/menus/menu_catalogos.py

from typing import Callable

from sqlalchemy.exc import IntegrityError

from book_manager.services.services import ServicioBase


class MenuCatalogos:
    """Menú CRUD genérico para entidades de catálogo."""

    def __init__(
        self,
        titulo: str,
        servicio: ServicioBase,
        crear_entidad: Callable,
        mostrar_entidad: Callable,
        modificar_entidad: Callable,
    ) -> None:
        self.titulo = titulo
        self.servicio = servicio
        self.crear_entidad = crear_entidad
        self.mostrar_entidad = mostrar_entidad
        self.modificar_entidad = modificar_entidad

    def ejecutar(self) -> None:
        """Ejecuta el menú CRUD del catálogo."""

        while True:
            print("\n" + "=" * 40)
            print(f"         GESTIÓN DE {self.titulo.upper()}")
            print("=" * 40)
            print(f"1. Listar {self.titulo.lower()}")
            print(f"2. Buscar {self.titulo.lower()}")
            print(f"3. Crear {self.titulo.lower()}")
            print(f"4. Modificar {self.titulo.lower()}")
            print(f"5. Eliminar {self.titulo.lower()}")
            print("0. Volver")
            print("=" * 40)

            opcion = input("Seleccione una opción: ").strip()

            if opcion == "1":
                self.listar()
            elif opcion == "2":
                self.buscar()
            elif opcion == "3":
                self.crear()
            elif opcion == "4":
                self.modificar()
            elif opcion == "5":
                self.eliminar()
            elif opcion == "0":
                break
            else:
                print("\nOpción inválida.")

    def listar(self) -> None:
        """Lista todas las entidades del catálogo."""

        entidades = self.servicio.obtener_todos()

        print(f"\n--- LISTADO DE {self.titulo.upper()} ---")

        if not entidades:
            print("No hay registros.")
            return

        for entidad in entidades:
            self.mostrar_entidad(entidad)

    def buscar(self) -> None:
        """Busca una entidad por ID."""

        try:
            entidad_id = int(input("Ingrese el ID: "))
            entidad = self.servicio.obtener_por_id(entidad_id)

            if entidad is None:
                print("\nRegistro no encontrado.")
                return

            print("\n--- REGISTRO ENCONTRADO ---")
            self.mostrar_entidad(entidad)

        except ValueError:
            print("\nEl ID debe ser numérico.")

    def crear(self) -> None:
        """Crea una nueva entidad."""

        try:
            entidad = self.crear_entidad()
            self.servicio.crear(entidad)
            print("\nRegistro creado correctamente.")

        except ValueError as error:
            print(f"\nError: {error}")

        except IntegrityError:
            print(
                "\nNo se pudo crear el registro. "
                "Verifique que los datos no estén duplicados "
                "y que las referencias sean válidas."
            )

    def modificar(self) -> None:
        """Modifica una entidad existente."""

        try:
            entidad_id = int(input("Ingrese el ID a modificar: "))
            entidad = self.servicio.obtener_por_id(entidad_id)

            if entidad is None:
                print("\nRegistro no encontrado.")
                return

            entidad_modificada = self.modificar_entidad(entidad)
            self.servicio.actualizar(entidad_modificada)

            print("\nRegistro modificado correctamente.")

        except ValueError as error:
            print(f"\nError: {error}")

        except IntegrityError:
            print(
                "\nNo se pudo modificar el registro. "
                "La operación incumple una restricción "
                "de la base de datos."
            )

    def eliminar(self) -> None:
        """Elimina una entidad por ID."""

        try:
            entidad_id = int(input("Ingrese el ID a eliminar: "))
            entidad = self.servicio.obtener_por_id(entidad_id)

            if entidad is None:
                print("\nRegistro no encontrado.")
                return

            confirmacion = input(
                "¿Confirma la eliminación? (s/n): "
            ).strip().lower()

            if confirmacion == "s":
                self.servicio.eliminar(entidad_id)
                print("\nRegistro eliminado correctamente.")
            else:
                print("\nOperación cancelada.")

        except ValueError as error:
            print(f"\nError: {error}")

        except IntegrityError:
            print(
                "\nNo se pudo eliminar el registro. "
                "Puede estar relacionado con otros registros."
            )
````

````python
# Modifiación console
%%writefile /content/Seminario_de_Actualizacion_1/book_manager/src/book_manager/ui/console.py

"""
Interfaz principal de consola del sistema Book Manager.
"""

import csv
from datetime import datetime
from decimal import Decimal, ROUND_HALF_UP
from pathlib import Path

from book_manager.ui.menus.catalogos_config import (
    crear_editorial,
    crear_genero,
    crear_moneda,
    crear_tipo_cotizacion,
    modificar_editorial,
    modificar_genero,
    modificar_moneda,
    modificar_tipo_cotizacion,
    mostrar_editorial,
    mostrar_genero,
    mostrar_moneda,
    mostrar_tipo_cotizacion,
)
from book_manager.ui.menus.menu_catalogos import MenuCatalogos
from book_manager.ui.menus.menu_cotizaciones import MenuCotizaciones
from book_manager.ui.menus.menu_libros import MenuLibros
from book_manager.ui.menus.menu_precios import MenuPrecios
from book_manager.ui.menus.menu_stock import MenuStock


class Consola:
    """Interfaz principal de usuario para Book Manager."""

    def __init__(
        self,
        servicio_libros,
        servicio_editoriales,
        servicio_generos,
        servicio_precios,
        servicio_monedas,
        servicio_stock,
        servicio_cotizaciones,
        servicio_tipos_cotizacion,
    ) -> None:

        self.servicio_libros = servicio_libros
        self.servicio_precios = servicio_precios
        self.servicio_monedas = servicio_monedas
        self.servicio_cotizaciones = servicio_cotizaciones
        self.servicio_tipos_cotizacion = servicio_tipos_cotizacion

        self.menu_libros = MenuLibros(
            servicio_libros,
            servicio_editoriales,
            servicio_generos,
        )

        self.menu_precios = MenuPrecios(
            servicio_precios,
            servicio_libros,
            servicio_monedas,
        )

        self.menu_stock = MenuStock(
            servicio_stock,
            servicio_libros,
        )

        self.menu_cotizaciones = MenuCotizaciones(
            servicio_cotizaciones,
            servicio_tipos_cotizacion,
            servicio_monedas,
        )

        self.menu_generos = MenuCatalogos(
            titulo="Géneros",
            servicio=servicio_generos,
            crear_entidad=crear_genero,
            mostrar_entidad=mostrar_genero,
            modificar_entidad=modificar_genero,
        )

        self.menu_editoriales = MenuCatalogos(
            titulo="Editoriales",
            servicio=servicio_editoriales,
            crear_entidad=crear_editorial,
            mostrar_entidad=mostrar_editorial,
            modificar_entidad=modificar_editorial,
        )

        self.menu_monedas = MenuCatalogos(
            titulo="Monedas",
            servicio=servicio_monedas,
            crear_entidad=crear_moneda,
            mostrar_entidad=mostrar_moneda,
            modificar_entidad=modificar_moneda,
        )

        self.menu_tipos_cotizacion = MenuCatalogos(
            titulo="Tipos de cotización",
            servicio=servicio_tipos_cotizacion,
            crear_entidad=crear_tipo_cotizacion,
            mostrar_entidad=mostrar_tipo_cotizacion,
            modificar_entidad=modificar_tipo_cotizacion,
        )

    # ======================================================
    # OPCIÓN 9 - OBTENER COTIZACIONES POR API
    # ======================================================

    def obtener_cotizaciones_api(self) -> None:
        """Actualiza cotizaciones desde la API y muestra los valores."""

        print("\n=== OBTENER COTIZACIONES POR API ===")
        print("Consultando cotizaciones del dólar...")

        try:
            resultado = self.servicio_cotizaciones.obtener_cotizaciones()

            print("\nCotizaciones procesadas correctamente.")
            print(f"Creadas: {resultado['creadas']}")
            print(f"Actualizadas: {resultado['actualizadas']}")
            print(f"Omitidas: {resultado['omitidas']}")

        except Exception as error:
            print(f"\nError al consultar la API: {error}")
            return

        # Recuperar las últimas cotizaciones guardadas.
        tipos = self.servicio_tipos_cotizacion.obtener_todos()

        print("\n=== ÚLTIMAS COTIZACIONES REGISTRADAS ===")

        print(
            "{:<22} {:<14} {:>15} {:>15}".format(
                "TIPO",
                "FECHA",
                "COMPRA",
                "VENTA",
            )
        )

        print("-" * 70)

        cantidad = 0

        for tipo in tipos:

            try:
                historico = (
                    self.servicio_cotizaciones.obtener_historico(
                        tipo.id
                    )
                )

                if not historico:
                    continue

                ultima = max(
                    historico,
                    key=lambda cotizacion: cotizacion.fecha,
                )

                compra = (
                    f"{ultima.valor_compra:.2f}"
                    if ultima.valor_compra is not None
                    else "N/D"
                )

                venta = (
                    f"{ultima.valor_venta:.2f}"
                    if ultima.valor_venta is not None
                    else "N/D"
                )

                print(
                    "{:<22} {:<14} {:>15} {:>15}".format(
                        tipo.nombre[:22],
                        str(ultima.fecha),
                        compra,
                        venta,
                    )
                )

                cantidad += 1

            except Exception as error:
                print(
                    f"No se pudo consultar {tipo.nombre}: {error}"
                )

        if cantidad == 0:
            print("\nNo hay cotizaciones registradas.")

        print(f"\nTipos de cotización mostrados: {cantidad}")

    # ======================================================
    # MÉTODOS AUXILIARES
    # ======================================================

    @staticmethod
    def redondear(valor: Decimal) -> Decimal:
        """Redondea importes a dos decimales."""

        return valor.quantize(
            Decimal("0.01"),
            rounding=ROUND_HALF_UP,
        )

    def obtener_ultima_cotizacion(self, tipo_id: int):
        """Recupera la cotización más reciente de un tipo."""

        cotizaciones = (
            self.servicio_cotizaciones.obtener_historico(tipo_id)
        )

        if not cotizaciones:
            return None

        return max(
            cotizaciones,
            key=lambda cotizacion: cotizacion.fecha,
        )

    def seleccionar_tipo_cotizacion(self):
        """Permite seleccionar un tipo de cotización."""

        tipos = self.servicio_tipos_cotizacion.obtener_todos()

        if not tipos:
            print("\nNo hay tipos de cotización registrados.")
            return None

        print("\nTipos de cotización disponibles:")

        for tipo in tipos:
            print(f"{tipo.id}. {tipo.nombre}")

        try:
            tipo_id = int(
                input(
                    "\nSeleccione el ID del tipo de cotización: "
                ).strip()
            )
        except ValueError:
            print("\nDebe ingresar un número entero.")
            return None

        tipo = next(
            (t for t in tipos if t.id == tipo_id),
            None,
        )

        if tipo is None:
            print("\nEl tipo de cotización no existe.")
            return None

        cotizacion = self.obtener_ultima_cotizacion(tipo_id)

        if cotizacion is None:
            print(
                "\nNo hay cotizaciones registradas para ese tipo."
            )
            return None

        return tipo, cotizacion

    def seleccionar_moneda(self):
        """
        Permite seleccionar una moneda por:
        - ID
        - Código
        - Nombre completo
        """

        monedas = self.servicio_monedas.obtener_todos()

        if not monedas:
            print("\nNo hay monedas registradas.")
            return None

        print("\nMonedas disponibles:")

        for moneda in monedas:
            print(
                f"{moneda.id}. "
                f"{moneda.codigo} - {moneda.nombre}"
            )

        seleccion = input(
            "\nIngrese el ID, código o nombre de la moneda: "
        ).strip().casefold()

        moneda_seleccionada = next(
            (
                moneda
                for moneda in monedas
                if str(moneda.id) == seleccion
                or moneda.codigo.strip().casefold() == seleccion
                or moneda.nombre.strip().casefold() == seleccion
            ),
            None,
        )

        if moneda_seleccionada is None:
            print("\nLa moneda seleccionada no existe.")
            return None

        return moneda_seleccionada

    # ======================================================
    # OPCIÓN 10 - PRECIOS BIMONETARIOS
    # ======================================================

    def ver_precios_bimonetarios(self) -> None:
        """
        Muestra precios en ARS y otra moneda seleccionada.

        Para USD calcula conversiones utilizando la
        cotización de venta seleccionada.

        Para otras monedas muestra los precios registrados.
        """

        print("\n=== PRECIOS BIMONETARIOS ===")

        moneda = self.seleccionar_moneda()

        if moneda is None:
            return

        codigo = moneda.codigo.upper()

        if codigo == "ARS":
            print(
                "\nSeleccione una moneda distinta de ARS."
            )
            return

        valor_dolar = None

        # Solicitar cotización solamente para USD.
        if codigo == "USD":

            seleccion = self.seleccionar_tipo_cotizacion()

            if seleccion is None:
                return

            tipo, cotizacion = seleccion

            if cotizacion.valor_venta is None:
                print("\nLa cotización de venta no está disponible.")
                return

            valor_dolar = Decimal(
                str(cotizacion.valor_venta)
            )

            if valor_dolar <= 0:
                print(
                    "\nLa cotización seleccionada no es válida."
                )
                return

            print(
                f"\nTipo: {tipo.nombre}"
                f" | Fecha: {cotizacion.fecha}"
                f" | Venta: {valor_dolar:.2f} ARS/USD"
            )

        precios = self.servicio_precios.obtener_todos()

        if not precios:
            print("\nNo hay precios registrados.")
            return

        # Agrupar precios por libro y moneda.
        precios_por_libro = {}

        for precio in precios:

            libro_id = precio.libro_id

            if libro_id not in precios_por_libro:
                precios_por_libro[libro_id] = {
                    "titulo": precio.libro.titulo,
                    "monedas": {},
                }

            codigo_origen = precio.moneda.codigo.upper()

            precios_por_libro[libro_id]["monedas"][
                codigo_origen
            ] = Decimal(str(precio.valor))

        print(
            "\n{:<35} {:>16} {:>16}".format(
                "LIBRO",
                "PRECIO ARS",
                f"PRECIO {codigo}",
            )
        )

        print("-" * 70)

        for datos in precios_por_libro.values():

            precio_ars = datos["monedas"].get("ARS")
            precio_destino = datos["monedas"].get(codigo)

            # Completar precios faltantes mediante conversión.
            if codigo == "USD" and valor_dolar is not None:

                if (
                    precio_ars is not None
                    and precio_destino is None
                ):
                    precio_destino = (
                        precio_ars / valor_dolar
                    )

                elif (
                    precio_destino is not None
                    and precio_ars is None
                ):
                    precio_ars = (
                        precio_destino * valor_dolar
                    )

            texto_ars = (
                f"{self.redondear(precio_ars):.2f}"
                if precio_ars is not None
                else "Sin precio"
            )

            texto_destino = (
                f"{self.redondear(precio_destino):.2f}"
                if precio_destino is not None
                else "Sin precio"
            )

            print(
                "{:<35} {:>16} {:>16}".format(
                    datos["titulo"][:35],
                    texto_ars,
                    texto_destino,
                )
            )

        if codigo == "USD":
            print(
                "\nNota: los precios faltantes en ARS/USD "
                "se calcularon usando la cotización seleccionada."
            )
        else:
            print(
                "\nNota: para esta moneda se muestran únicamente "
                "los precios registrados en la base de datos."
            )

    # ======================================================
    # OPCIÓN 11 - EXPORTAR PRECIOS A CSV
    # ======================================================
    def exportar_precios_csv(self) -> None:
        """
        Exporta todos los precios registrados en la base
        de datos, respetando la moneda original.
        """

        print("\n=== EXPORTAR PRECIOS A CSV ===")

        precios = self.servicio_precios.obtener_todos()

        if not precios:
            print("\nNo hay precios registrados para exportar.")
            return

        # Preparar los datos del archivo
        filas = []

        for precio in precios:
            filas.append({
                "libro_id": precio.libro_id,
                "isbn": precio.libro.isbn,
                "titulo": precio.libro.titulo,
                "moneda": precio.moneda.codigo,
                "precio": str(precio.valor),
            })

        # Definir la ubicación del archivo CSV
        ruta_proyecto = Path(__file__).resolve().parents[3]
        carpeta_exportaciones = ruta_proyecto / "exportaciones"

        try:
            carpeta_exportaciones.mkdir(
                parents=True,
                exist_ok=True,
            )

            fecha_archivo = datetime.now().strftime(
                "%Y%m%d_%H%M%S"
            )

            ruta_csv = (
                carpeta_exportaciones
                / f"precios_{fecha_archivo}.csv"
            )

            columnas = [
                "libro_id",
                "isbn",
                "titulo",
                "moneda",
                "precio",
            ]

            with open(
                ruta_csv,
                mode="w",
                newline="",
                encoding="utf-8-sig",
            ) as archivo:

                escritor = csv.DictWriter(
                    archivo,
                    fieldnames=columnas,
                )

                escritor.writeheader()
                escritor.writerows(filas)

            print("\nArchivo CSV generado correctamente.")
            print(f"Registros exportados: {len(filas)}")
            print(f"Ubicación: {ruta_csv}")

        except OSError as error:
            print(f"\nError al exportar el archivo: {error}")


    # ======================================================
    # MENÚ PRINCIPAL
    # ======================================================

    def ejecutar(self) -> None:
        """Ejecuta el menú principal de la aplicación."""

        while True:

            self.mostrar_menu_principal()

            opcion = input(
                "Seleccione una opción: "
            ).strip()

            if opcion == "1":
                self.menu_libros.ejecutar()

            elif opcion == "2":
                self.menu_precios.ejecutar()

            elif opcion == "3":
                self.menu_stock.ejecutar()

            elif opcion == "4":
                self.menu_cotizaciones.ejecutar()

            elif opcion == "5":
                self.menu_generos.ejecutar()

            elif opcion == "6":
                self.menu_editoriales.ejecutar()

            elif opcion == "7":
                self.menu_monedas.ejecutar()

            elif opcion == "8":
                self.menu_tipos_cotizacion.ejecutar()

            elif opcion == "9":
                self.obtener_cotizaciones_api()

            elif opcion == "10":
                self.ver_precios_bimonetarios()

            elif opcion == "11":
                self.exportar_precios_csv()

            elif opcion == "0":
                print("\nSaliendo de Book Manager...")
                break

            else:
                print(
                    "\nOpción inválida. Intente nuevamente."
                )

    @staticmethod
    def mostrar_menu_principal() -> None:
        """Muestra las opciones principales del sistema."""

        print("\n" + "=" * 45)
        print("              BOOK MANAGER")
        print("=" * 45)

        print("1. Gestión de libros")
        print("2. Gestión de precios")
        print("3. Gestión de stock")
        print("4. Gestión de cotizaciones")
        print("5. Gestión de géneros")
        print("6. Gestión de editoriales")
        print("7. Gestión de monedas")
        print("8. Gestión de tipos de cotización")
        print("9. Obtener cotizaciones por API")
        print("10. Ver precios bimonetarios")
        print("11. Exportar precios a CSV")
        print("0. Salir")

        print("=" * 45)
````

````python
# Creación de tablas en Supabase

import os
import sys
import importlib
from pathlib import Path

from google.colab import userdata
from sqlalchemy import create_engine, inspect, text
from sqlalchemy.engine import make_url


# --------------------------------------------------
# 1. Configurar la ruta del proyecto
# --------------------------------------------------

RUTA_SRC = Path(
    "/content/Seminario_de_Actualizacion_1/book_manager/src"
)

if not RUTA_SRC.exists():
    raise FileNotFoundError(
        "No se encontró el proyecto. Verificá que hayas clonado el repositorio."
    )

sys.path = [p for p in sys.path if p != str(RUTA_SRC)]
sys.path.insert(0, str(RUTA_SRC))

# Limpiar módulos anteriores
for nombre in list(sys.modules):
    if nombre == "book_manager" or nombre.startswith("book_manager."):
        del sys.modules[nombre]

importlib.invalidate_caches()

from book_manager.models.models import Base


# --------------------------------------------------
# 2. Recuperar la conexión a Supabase
# --------------------------------------------------

database_url = userdata.get("DATABASE_URL")

if not database_url:
    raise ValueError(
        "No se encontró DATABASE_URL en los secretos de Colab."
    )

os.environ["DATABASE_URL"] = database_url

url = make_url(database_url)
url = url.set(drivername="postgresql+psycopg")


# --------------------------------------------------
# 3. Crear el motor de conexión
# --------------------------------------------------

engine = create_engine(
    url,
    pool_pre_ping=True,
    connect_args={"sslmode": "require"}
)

assert engine.dialect.name == "postgresql"

with engine.connect() as conexion:
    print(
        "Base de datos:",
        conexion.execute(
            text("SELECT current_database()")
        ).scalar()
    )


# --------------------------------------------------
# 4. Verificar y crear las tablas
# --------------------------------------------------

inspector = inspect(engine)

tablas_existentes = set(inspector.get_table_names())
tablas_modelo = set(Base.metadata.tables.keys())

tablas_faltantes = tablas_modelo - tablas_existentes

if not tablas_faltantes:

    print("\nLas tablas ya fueron creadas con anterioridad.")

else:

    Base.metadata.create_all(bind=engine)

    print("\nTablas creadas correctamente.")

    print("Tablas nuevas:")
    for tabla in sorted(tablas_faltantes):
        print(f" - {tabla}")


# --------------------------------------------------
# 5. Mostrar el estado de las tablas
# --------------------------------------------------

print("\n=== TABLAS EN SUPABASE ===")

inspector = inspect(engine)

for tabla in sorted(Base.metadata.tables):
    existe = inspector.has_table(tabla)

    print(
        f"{tabla}: {'OK' if existe else 'NO ENCONTRADA'}"
    )
````

#### Migracion de datos

````python
# Migración de tablas: SQLite -> Supabase

from pathlib import Path
from sqlalchemy import MetaData, create_engine, select, func, text

# --------------------------------------------------
# 1. Conexiones a las bases de datos
# --------------------------------------------------

ruta_sqlite = Path(
    "/content/Seminario_de_Actualizacion_1/"
    "book_manager/book_manager.db"
)

if not ruta_sqlite.is_file():
    raise FileNotFoundError(
        f"No se encontró la base SQLite: {ruta_sqlite}"
    )

engine_sqlite = create_engine(
    f"sqlite:///{ruta_sqlite}"
)

# Recuperar la conexión a Supabase creada
# en la celda anterior.
if "engine" not in globals():
    raise RuntimeError(
        "No se encontró la conexión a Supabase. "
        "Ejecutá primero la celda de creación de tablas."
    )

if engine.dialect.name != "postgresql":
    raise RuntimeError(
        "La conexión de destino no es PostgreSQL."
    )

# --------------------------------------------------
# 2. Orden de migración según las dependencias
# --------------------------------------------------

tablas = [
    "generos",
    "editoriales",
    "monedas",
    "tipos_cotizacion",
    "libros",
    "precios",
    "stock",
    "cotizaciones_dolar"
]

metadata_sqlite = MetaData()
metadata_pg = MetaData()

metadata_sqlite.reflect(
    bind=engine_sqlite,
    only=tablas
)

metadata_pg.reflect(
    bind=engine,
    only=tablas
)

# --------------------------------------------------
# 3. Leer los datos originales de SQLite
# --------------------------------------------------

datos = {}
cantidades_sqlite = {}

with engine_sqlite.connect() as origen:

    for nombre in tablas:

        tabla = metadata_sqlite.tables[nombre]

        registros = origen.execute(
            select(tabla)
        ).mappings().all()

        datos[nombre] = [
            dict(registro) for registro in registros
        ]

        cantidades_sqlite[nombre] = len(registros)

# --------------------------------------------------
# 4. Verificar los datos existentes en Supabase
# --------------------------------------------------

cantidades_supabase = {}

with engine.connect() as destino:

    for nombre in tablas:

        tabla = metadata_pg.tables[nombre]

        cantidad = destino.execute(
            select(func.count()).select_from(tabla)
        ).scalar_one()

        cantidades_supabase[nombre] = cantidad

print("\n=== ESTADO DE LA MIGRACIÓN ===")

for nombre in tablas:

    print(
        f"{nombre}: "
        f"SQLite={cantidades_sqlite[nombre]} | "
        f"Supabase={cantidades_supabase[nombre]}"
    )

# --------------------------------------------------
# 5. Determinar si es necesario migrar
# --------------------------------------------------

hay_datos_supabase = any(
    cantidad > 0
    for cantidad in cantidades_supabase.values()
)

if hay_datos_supabase:

    cantidades_iguales = all(
        cantidades_sqlite[nombre] == cantidades_supabase[nombre]
        for nombre in tablas
    )

    if cantidades_iguales:

        print(
            "\nLos datos ya fueron migrados con anterioridad."
        )
        print(
            "No es necesario realizar nuevamente la migración."
        )

    else:

        print(
            "\nATENCIÓN: Supabase ya contiene registros, "
            "pero las cantidades no coinciden con SQLite."
        )
        print(
            "No se realizará la migración automática "
            "para evitar duplicados o sobrescribir datos."
        )

else:

    # --------------------------------------------------
    # 6. Migrar los datos a Supabase
    # --------------------------------------------------

    with engine.begin() as destino:

        for nombre in tablas:

            registros = datos[nombre]

            if registros:

                destino.execute(
                    metadata_pg.tables[nombre].insert(),
                    registros
                )

            print(
                f"{nombre}: "
                f"{len(registros)} registros migrados"
            )

        # --------------------------------------------------
        # 7. Sincronizar secuencias de IDs
        # --------------------------------------------------

        for nombre in tablas:

            tabla = metadata_pg.tables[nombre]

            for columna in tabla.primary_key.columns:

                secuencia = destino.execute(
                    text(
                        "SELECT pg_get_serial_sequence("
                        ":tabla, :columna)"
                    ),
                    {
                        "tabla": f"public.{nombre}",
                        "columna": columna.name
                    }
                ).scalar()

                if secuencia:

                    nombre_columna = columna.name

                    # Identificadores provenientes del modelo,
                    # no de entradas del usuario.
                    nombre_tabla_sql = destino.dialect.identifier_preparer.quote(
                        nombre
                    )
                    nombre_columna_sql = destino.dialect.identifier_preparer.quote(
                        nombre_columna
                    )

                    max_id = destino.execute(
                        text(
                            f"SELECT MAX({nombre_columna_sql}) "
                            f"FROM {nombre_tabla_sql}"
                        )
                    ).scalar()

                    destino.execute(
                        text(
                            "SELECT setval("
                            "CAST(:secuencia AS regclass), "
                            ":valor, :tiene_registros)"
                        ),
                        {
                            "secuencia": secuencia,
                            "valor": max_id if max_id is not None else 1,
                            "tiene_registros": max_id is not None
                        }
                    )

    print("\nMigración finalizada correctamente.")
````
