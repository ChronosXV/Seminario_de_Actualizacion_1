### 02 Cargar datos de SQL

````python
# Creacion y modificacion del archivo preload, adicionamos esta funcion

from pathlib import Path

RUTA_ARCHIVO = Path(
    "/content/Seminario_de_Actualizacion_1/"
    "book_manager/src/book_manager/preload_data/preload_data.py"
)

codigo = RUTA_ARCHIVO.read_text(encoding="utf-8")

nueva_funcion = '''

def cargar_datos_desde_sql(
    conexion: ConexionDB,
    ruta_sql: str | Path,
) -> int:
    """
    Carga datos ejecutando instrucciones SQL desde un archivo.

    Args:
        conexion: Administrador de conexión a la base de datos.
        ruta_sql: Ruta del archivo SQL que se desea ejecutar.

    Returns:
        int: Cantidad de instrucciones SQL ejecutadas.

    Raises:
        FileNotFoundError: Si el archivo SQL no existe.
        ValueError: Si el archivo no contiene instrucciones válidas.
    """
    archivo = Path(ruta_sql)

    if not archivo.is_file():
        raise FileNotFoundError(
            f"No se encontró el archivo SQL: {archivo}"
        )

    contenido = archivo.read_text(encoding="utf-8")

    instrucciones = []

    for bloque in contenido.split(";"):
        lineas = [
            linea
            for linea in bloque.splitlines()
            if linea.strip()
            and not linea.strip().startswith("--")
        ]

        instruccion = "\\n".join(lineas).strip()

        if not instruccion:
            continue

        if instruccion.upper() in (
            "BEGIN",
            "BEGIN TRANSACTION",
            "COMMIT",
        ):
            continue

        instrucciones.append(instruccion)

    if not instrucciones:
        raise ValueError(
            "El archivo SQL no contiene instrucciones para ejecutar."
        )

    with conexion.obtener_sesion() as sesion:
        for instruccion in instrucciones:
            sesion.execute(text(instruccion))

    print(
        f"Carga SQL finalizada: "
        f"{len(instrucciones)} instrucciones ejecutadas."
    )

    return len(instrucciones)
'''

if "def cargar_datos_desde_sql(" not in codigo:
    RUTA_ARCHIVO.write_text(
        codigo.rstrip() + "\n" + nueva_funcion,
        encoding="utf-8"
    )
    print("Función cargar_datos_desde_sql() agregada correctamente.")
else:
    print("La función ya existe. No se realizaron modificaciones.")
````

````python
# @title Implementación del loaders.py

from pathlib import Path

ruta = Path(
    "/content/Seminario_de_Actualizacion_1/"
    "book_manager/src/book_manager/scraper/loaders.py"
)

codigo = '''"""
Procesamiento y limpieza de los datos extraídos con Scrapy.

Sprint 3 - Ejercicio 03.
"""

import re
from decimal import Decimal, InvalidOperation

from itemloaders import ItemLoader
from itemloaders.processors import Identity, Join, MapCompose, TakeFirst

from book_manager.scraper.items import LibroItem


def limpiar_texto(valor: str) -> str:
    """Elimina espacios innecesarios de un texto."""
    return " ".join(valor.split())


def limpiar_precio(valor: str) -> Decimal:
    """Convierte un precio argentino a Decimal."""
    texto = re.sub(r"[^0-9,.-]", "", str(valor).strip())

    if not texto:
        raise ValueError("El precio está vacío.")

    # Formato argentino: 15.000,50
    if "," in texto:
        texto = texto.replace(".", "").replace(",", ".")
    elif texto.count(".") > 1:
        texto = texto.replace(".", "")
    elif "." in texto:
        parte_entera, parte_decimal = texto.split(".")
        if len(parte_decimal) == 3:
            texto = parte_entera + parte_decimal

    try:
        return Decimal(texto)
    except InvalidOperation as error:
        raise ValueError(f"Precio inválido: {valor}") from error


class LibroLoader(ItemLoader):
    """Limpia y organiza la información extraída de cada libro."""

    default_item_class = LibroItem

    titulo_in = MapCompose(limpiar_texto)
    titulo_out = TakeFirst()

    precio_in = MapCompose(limpiar_precio)
    precio_out = TakeFirst()

    imagen_url_in = MapCompose(str.strip)
    imagen_url_out = TakeFirst()

    medios_pago_out = Identity()

    descripcion_in = MapCompose(limpiar_texto)
    descripcion_out = Join(" ")

    url_in = MapCompose(str.strip)
    url_out = TakeFirst()
'''

ruta.write_text(codigo, encoding="utf-8")

print("OK - loaders.py implementado correctamente.")
````

````python
# @title Implementacion de pipelines.py

from pathlib import Path

ruta = Path(
    "/content/Seminario_de_Actualizacion_1/"
    "book_manager/src/book_manager/scraper/pipelines.py"
)

codigo = '''"""
Validación de libros extraídos mediante Scrapy.

Sprint 3 - Ejercicio 03.
"""

from decimal import Decimal, InvalidOperation

from scrapy.exceptions import DropItem
from scrapy.item import Item


class ValidacionLibroPipeline:
    """Valida los datos obligatorios de cada libro extraído."""

    def process_item(self, item: Item) -> Item:
        """Valida el libro y lo devuelve si cumple las condiciones."""

        titulo = item.get("titulo")
        precio = item.get("precio")
        url = item.get("url")

        if not isinstance(titulo, str) or not titulo.strip():
            raise DropItem("Libro descartado: título vacío.")

        if not isinstance(url, str) or not url.strip():
            raise DropItem("Libro descartado: URL vacía.")

        if precio is None:
            raise DropItem("Libro descartado: precio inexistente.")

        try:
            precio_decimal = Decimal(str(precio))
        except (InvalidOperation, ValueError, TypeError) as error:
            raise DropItem(
                f"Libro descartado: precio inválido ({precio})."
            ) from error

        if not precio_decimal.is_finite() or precio_decimal <= 0:
            raise DropItem(
                "Libro descartado: el precio debe ser positivo."
            )

        item["precio"] = precio_decimal

        return item
'''

# Comprobar sintaxis antes de guardar
compile(codigo, str(ruta), "exec")

ruta.write_text(codigo, encoding="utf-8")

print("OK - pipelines.py actualizado correctamente.")
````

````python
# @title Implementación de CuspideSpider

from pathlib import Path

ruta = Path(
    "/content/Seminario_de_Actualizacion_1/"
    "book_manager/src/book_manager/scraper/spiders/cuspide_spider.py"
)

codigo = r'''
"""
Spider para buscar libros del catálogo de Book Manager en Cúspide.

Sprint 3 - Ejercicio 03.
"""

import json
import re
from urllib.parse import urlencode

import scrapy

from book_manager.scraper.loaders import LibroLoader


class CuspideSpider(scrapy.Spider):
    """Busca hasta diez resultados de Cúspide por cada título recibido."""

    name = "cuspide"

    allowed_domains = ["cuspide.com"]

    custom_settings = {
        "ROBOTSTXT_OBEY": True,
        "DOWNLOAD_DELAY": 1,
        "CONCURRENT_REQUESTS_PER_DOMAIN": 1,
        "ITEM_PIPELINES": {
            "book_manager.scraper.pipelines.ValidacionLibroPipeline": 300
        },
    }

    def __init__(self, titulos_json=None, *args, **kwargs):
        """Recibe los títulos de los libros de nuestro catálogo."""

        super().__init__(*args, **kwargs)

        if not titulos_json:
            raise ValueError(
                "Debe proporcionar los títulos del catálogo."
            )

        titulos = json.loads(titulos_json)

        if not isinstance(titulos, list):
            raise ValueError(
                "Los títulos deben proporcionarse como una lista JSON."
            )

        self.titulos = [
            titulo.strip()
            for titulo in titulos
            if isinstance(titulo, str) and titulo.strip()
        ]

        if not self.titulos:
            raise ValueError(
                "No se recibieron títulos válidos."
            )

    async def start(self):
        """Genera una búsqueda por cada título del catálogo."""

        for titulo in self.titulos:

            parametros = urlencode({
                "s": titulo,
                "post_type": "product"
            })

            url_busqueda = (
                f"https://cuspide.com/?{parametros}"
            )

            yield scrapy.Request(
                url=url_busqueda,
                callback=self.parse,
                cb_kwargs={
                    "titulo_buscado": titulo
                },
                dont_filter=True
            )

    def parse(self, response, titulo_buscado):
        """Obtiene hasta diez resultados de una búsqueda por título."""

        productos = response.css("div.product")

        encontrados = 0

        for producto in productos:

            titulo = producto.css(
                ".woocommerce-loop-product__title"
            ).xpath("string(.)").get()

            if not titulo:
                continue

            titulo = titulo.strip()

            # Excluir tarjetas de regalo
            if "GIFT CARD" in titulo.upper():
                continue

            url = producto.css(
                "a.woocommerce-LoopProduct-link::attr(href)"
            ).get()

            if not url:
                continue

            yield response.follow(
                url,
                callback=self.parse_libro,
                cb_kwargs={
                    "titulo_buscado": titulo_buscado
                }
            )

            encontrados += 1

            if encontrados >= 10:
                break

    def parse_libro(self, response, titulo_buscado):
        """Extrae los datos de cada resultado encontrado."""

        loader = LibroLoader(response=response)

        # ------------------------------------------
        # Título
        # ------------------------------------------

        titulo = response.css(
            "h1.product_title"
        ).xpath("string(.)").get()

        # ------------------------------------------
        # Precio
        # ------------------------------------------

        precio = response.css(
            ".summary .price"
        ).xpath("string(.)").get()

        # ------------------------------------------
        # Descripción
        # ------------------------------------------

        descripcion = response.css(
            ".description"
        ).xpath("string(.)").get()

        # Si no existe descripción HTML, buscar en JSON-LD
        if not descripcion or not descripcion.strip():

            scripts = response.css(
                'script[type="application/ld+json"]::text'
            ).getall()

            for script in scripts:

                try:
                    datos = json.loads(script)

                except json.JSONDecodeError:
                    continue

                descripcion = self.buscar_descripcion(datos)

                if descripcion:
                    break

        # ------------------------------------------
        # Imagen del libro
        # ------------------------------------------

        imagen = response.css(
            "div.product img::attr(src)"
        ).get()

        # ------------------------------------------
        # Cargar los datos mediante ItemLoader
        # ------------------------------------------

        if titulo:
            loader.add_value(
                "titulo",
                titulo
            )

        if precio:
            loader.add_value(
                "precio",
                precio
            )

        # Limpiar la descripción antes de cargarla
        if descripcion:

            descripcion = self.limpiar_descripcion(
                descripcion,
                titulo
            )

            if descripcion:
                loader.add_value(
                    "descripcion",
                    descripcion
                )

        if imagen:
            loader.add_value(
                "imagen_url",
                response.urljoin(imagen)
            )

        loader.add_value(
            "url",
            response.url
        )

        # ------------------------------------------
        # Formas de pago
        # ------------------------------------------

        medios_pago = []

        for lista in response.css("ul"):

            opciones = lista.xpath("./li")

            textos = [
                " ".join(
                    opcion.xpath("string(.)").get(default="").split()
                )
                for opcion in opciones
            ]

            # Identificar el listado de medios de pago
            if any(
                "depósito o transferencia" in texto.lower()
                for texto in textos
            ):

                for texto in textos:

                    forma_pago = texto.lstrip("> ").strip()

                    if forma_pago:
                        medios_pago.append({
                            "forma_pago": forma_pago,
                            "precio": None
                        })

                break

        # Incorporar las formas de pago al ItemLoader
        if medios_pago:
            loader.add_value(
                "medios_pago",
                medios_pago
            )

        # ------------------------------------------
        # Construir el Item
        # ------------------------------------------

        item = loader.load_item()

        # Asociar el resultado al título de nuestro catálogo
        item["titulo_buscado"] = titulo_buscado

        # Enviar el resultado al Pipeline
        yield item

    @staticmethod
    def limpiar_descripcion(descripcion, titulo):
        """
        Limpia la sinopsis del producto.

        Elimina:
        - Encabezado 'Sinopsis del producto'.
        - Título repetido al final.
        - Posible ISBN-13 al final.
        """

        texto = descripcion.strip()

        # Eliminar encabezado
        texto = re.sub(
            r"^Sinopsis del producto\s*",
            "",
            texto,
            flags=re.IGNORECASE
        )

        # Normalizar espacios del título
        titulo_limpio = (
            " ".join(titulo.split())
            if titulo
            else ""
        )

        # Eliminar título repetido al final
        if titulo_limpio:

            patron_titulo = re.escape(titulo_limpio)

            texto = re.sub(
                rf"\s*{patron_titulo}\s*$",
                "",
                texto,
                flags=re.IGNORECASE
            ).rstrip()

        # Eliminar posible ISBN-13 al final
        texto = re.sub(
            r"(?<!\d)97[89]\d{10}$",
            "",
            texto
        ).rstrip()

        return texto

    @staticmethod
    def buscar_descripcion(datos):
        """
        Busca una descripción dentro de una estructura JSON-LD.

        Admite diccionarios y listas anidadas.
        """

        if isinstance(datos, dict):

            descripcion = datos.get("description")

            if isinstance(descripcion, str) and descripcion.strip():
                return descripcion

            for valor in datos.values():

                encontrada = CuspideSpider.buscar_descripcion(
                    valor
                )

                if encontrada:
                    return encontrada

        elif isinstance(datos, list):

            for elemento in datos:

                encontrada = CuspideSpider.buscar_descripcion(
                    elemento
                )

                if encontrada:
                    return encontrada

        return None
'''

# Crear la carpeta si no existe
ruta.parent.mkdir(
    parents=True,
    exist_ok=True
)

# Verificar la sintaxis antes de guardar
compile(codigo, str(ruta), "exec")

# Guardar el Spider
ruta.write_text(
    codigo,
    encoding="utf-8"
)

print("OK - CuspideSpider.")
````

````python
# @title Ejercicio 04 - Scraping, conversión y alertas CSV

# ============================================================
# CONFIGURACIÓN E IMPORTACIÓN DEL MÓDULO
# ============================================================

import os
import sys
import importlib
from pathlib import Path

from google.colab import userdata, files

# Ruta del proyecto
PROYECTO = Path(
    "/content/Seminario_de_Actualizacion_1/book_manager"
)

SRC = PROYECTO / "src"

RUTA_MODULO = (
    SRC / "book_manager/scraper/ejecutar_scraper.py"
)

# Verificar que el archivo exista
if not RUTA_MODULO.is_file():
    raise FileNotFoundError(
        f"No se encontró el módulo: {RUTA_MODULO}"
    )

# Priorizar la ruta correcta
if str(SRC) in sys.path:
    sys.path.remove(str(SRC))

sys.path.insert(0, str(SRC))

# Limpiar importaciones anteriores del paquete
for nombre in list(sys.modules):
    if nombre == "book_manager" or nombre.startswith("book_manager."):
        del sys.modules[nombre]

importlib.invalidate_caches()

# Importar el módulo
from book_manager.scraper.ejecutar_scraper import main

# ============================================================
# 1. CONFIGURACIÓN DEL PROYECTO
# ============================================================

PROYECTO = Path(
    "/content/Seminario_de_Actualizacion_1/book_manager"
)

SRC = PROYECTO / "src"

RUTA_MODULO = (
    SRC / "book_manager/scraper/ejecutar_scraper.py"
)

RUTA_SPIDER = (
    SRC / "book_manager/scraper/spiders/cuspide_spider.py"
)

if not RUTA_SPIDER.is_file():
    raise FileNotFoundError(
        f"No se encontró el Spider: {RUTA_SPIDER}"
    )

database_url = userdata.get("DATABASE_URL")

if not database_url:
    raise ValueError(
        "No se encontró DATABASE_URL en los secretos de Colab."
    )


# ============================================================
# 2. CÓDIGO DEL MÓDULO ejecutar_scraper.py
# ============================================================

codigo = r'''
"""
Ejercicio 04 - Book Manager.

Ejecuta el scraping del catálogo, convierte precios USD a ARS
y genera un reporte CSV con comparaciones y alertas.
"""

import csv
import json
import os
import re
import subprocess
import sys
import unicodedata

from datetime import datetime
from decimal import Decimal, InvalidOperation, ROUND_HALF_UP
from pathlib import Path

from sqlalchemy import select

from book_manager.database.connection import ConexionDB
from book_manager.models.models import (
    LibroModel,
    PrecioModel,
    MonedaModel,
    CotizacionDolarModel,
    TipoCotizacionModel,
)


# ============================================================
# COLUMNAS DEL REPORTE
# ============================================================

COLUMNAS_CSV = [
    "fecha_ejecucion",
    "libro_id",
    "titulo",
    "precio_original",
    "moneda_original",
    "cotizacion_oficial",
    "fecha_cotizacion",
    "precio_interno_ars",
    "precio_web_ars",
    "diferencia_ars",
    "diferencia_porcentaje",
    "umbral_porcentaje",
    "estado",
    "url",
]


# ============================================================
# NORMALIZACIÓN DE TÍTULOS
# ============================================================

def normalizar_titulo(titulo):
    """
    Normaliza títulos para comparar sin distinguir
    mayúsculas, tildes, puntuación ni espacios extra.
    """

    texto = unicodedata.normalize(
        "NFKD",
        titulo or ""
    )

    texto = "".join(
        caracter
        for caracter in texto
        if not unicodedata.combining(caracter)
    )

    texto = re.sub(
        r"[^a-z0-9]+",
        " ",
        texto.casefold()
    )

    return " ".join(texto.split())


# ============================================================
# SOLICITAR PORCENTAJE DE ALERTA
# ============================================================

def solicitar_umbral():
    """
    Solicita al usuario el porcentaje mínimo
    de diferencia para generar una alerta.
    """

    while True:

        entrada = input(
            "Ingresá el porcentaje de alerta: "
        ).strip().replace(",", ".")

        try:

            umbral = Decimal(entrada)

            if not umbral.is_finite() or umbral < 0:
                raise ValueError

            return umbral

        except (InvalidOperation, ValueError):

            print(
                "Ingresá un porcentaje válido "
                "mayor o igual a cero."
            )


# ============================================================
# OBTENER CATÁLOGO DESDE SUPABASE
# ============================================================

def obtener_catalogo(conexion):
    """
    Obtiene los libros y sus precios desde la base de datos.
    """

    consulta = (
        select(
            LibroModel.id,
            LibroModel.titulo,
            PrecioModel.valor,
            MonedaModel.codigo,
        )
        .join(
            PrecioModel,
            PrecioModel.libro_id == LibroModel.id,
        )
        .join(
            MonedaModel,
            MonedaModel.id == PrecioModel.moneda_id,
        )
        .order_by(
            LibroModel.id,
            PrecioModel.id,
        )
    )

    with conexion.obtener_sesion() as sesion:
        filas = sesion.execute(consulta).all()

    catalogo = [
        {
            "id": libro_id,
            "titulo": titulo,
            "precio_original": Decimal(str(precio)),
            "moneda": moneda.upper(),
        }
        for libro_id, titulo, precio, moneda in filas
    ]

    return catalogo


# ============================================================
# OBTENER COTIZACIÓN OFICIAL DEL DÓLAR
# ============================================================

def obtener_dolar_oficial(conexion):
    """
    Obtiene la última cotización oficial de venta
    del dólar registrada en Supabase.
    """

    consulta = (
        select(
            CotizacionDolarModel.fecha,
            CotizacionDolarModel.valor_venta,
        )
        .join(
            TipoCotizacionModel,
            CotizacionDolarModel.tipo_cotizacion_id
            == TipoCotizacionModel.id,
        )
        .join(
            MonedaModel,
            CotizacionDolarModel.moneda_id
            == MonedaModel.id,
        )
        .where(
            TipoCotizacionModel.nombre == "Oficial",
            MonedaModel.codigo == "USD",
        )
        .order_by(
            CotizacionDolarModel.fecha.desc(),
            CotizacionDolarModel.id.desc(),
        )
        .limit(1)
    )

    with conexion.obtener_sesion() as sesion:
        fila = sesion.execute(consulta).first()

    if fila is None:
        return None

    fecha, valor_venta = fila

    cotizacion = Decimal(str(valor_venta))

    if not cotizacion.is_finite() or cotizacion <= 0:
        return None

    return {
        "fecha": fecha,
        "venta": cotizacion,
    }


# ============================================================
# CONVERTIR PRECIOS A PESOS ARGENTINOS
# ============================================================

def convertir_precios(catalogo, dolar_oficial):
    """
    Convierte los precios USD a ARS utilizando
    la cotización oficial de venta.

    Los precios en ARS se conservan sin cambios.
    """

    for libro in catalogo:

        moneda = libro["moneda"]
        precio = libro["precio_original"]

        libro["precio_ars"] = None
        libro["cotizacion"] = None
        libro["fecha_cotizacion"] = None
        libro["estado_conversion"] = "OK"

        if moneda == "ARS":

            libro["precio_ars"] = precio

        elif moneda == "USD":

            if dolar_oficial is None:

                libro["estado_conversion"] = "SIN COTIZACIÓN"
                continue

            libro["cotizacion"] = dolar_oficial["venta"]

            libro["fecha_cotizacion"] = dolar_oficial["fecha"]

            libro["precio_ars"] = (
                precio * dolar_oficial["venta"]
            ).quantize(
                Decimal("0.01"),
                rounding=ROUND_HALF_UP,
            )

        else:

            libro["estado_conversion"] = (
                "MONEDA NO SOPORTADA"
            )

    return catalogo


# ============================================================
# EJECUTAR EL SPIDER
# ============================================================

def ejecutar_spider(proyecto, titulos, ruta_json):
    """
    Ejecuta CuspideSpider utilizando los títulos
    obtenidos desde el catálogo de Supabase.
    """

    ruta_spider = (
        proyecto
        / "src/book_manager/scraper/spiders/cuspide_spider.py"
    )

    entorno = os.environ.copy()

    entorno["PYTHONPATH"] = (
        str(proyecto / "src")
        + os.pathsep
        + entorno.get("PYTHONPATH", "")
    )

    comando = [
        sys.executable,
        "-m",
        "scrapy",
        "runspider",
        str(ruta_spider),
        "-a",
        "titulos_json=" + json.dumps(
            titulos,
            ensure_ascii=False,
        ),
        "-O",
        str(ruta_json),
        "-s",
        "LOG_LEVEL=ERROR",
    ]

    print(
        f"\nEjecutando Spider para {len(titulos)} títulos..."
    )

    resultado = subprocess.run(
        comando,
        cwd=proyecto,
        env=entorno,
        capture_output=True,
        text=True,
    )

    if resultado.returncode != 0:

        print(resultado.stderr[-5000:])

        raise RuntimeError(
            "No se pudo completar la ejecución del Spider."
        )

    if not ruta_json.is_file():

        raise FileNotFoundError(
            "No se generó el archivo JSON del scraping."
        )

    with ruta_json.open(
        "r",
        encoding="utf-8"
    ) as archivo:

        datos = json.load(archivo)

    if not isinstance(datos, list):

        raise ValueError(
            "El JSON del scraping no contiene una lista."
        )

    return datos


# ============================================================
# GENERAR COMPARACIONES Y ALERTAS
# ============================================================

def generar_reporte(catalogo, resultados_web, umbral, fecha):
    """
    Compara los precios internos con los precios web.

    Incluye:
    - ALERTA
    - SIN ALERTA
    - SIN COINCIDENCIA
    - SIN COTIZACIÓN
    - MONEDA NO SOPORTADA
    - SIN PRECIO WEB VÁLIDO
    """

    resultados_por_busqueda = {}

    for resultado in resultados_web:

        clave = normalizar_titulo(
            resultado.get("titulo_buscado", "")
        )

        resultados_por_busqueda.setdefault(
            clave,
            []
        ).append(resultado)

    reporte = []

    for libro in catalogo:

        titulo = libro["titulo"]

        clave = normalizar_titulo(titulo)

        coincidencias = [
            resultado
            for resultado in resultados_por_busqueda.get(
                clave,
                []
            )
            if normalizar_titulo(
                resultado.get("titulo", "")
            ) == clave
        ]

        # Evitar productos duplicados por URL.
        coincidencias_unicas = {}

        for resultado in coincidencias:

            url = resultado.get("url", "")

            if url:
                coincidencias_unicas[url] = resultado

        coincidencias = list(
            coincidencias_unicas.values()
        )

        # Datos comunes a todas las filas del libro.
        fila_base = {
            "fecha_ejecucion": fecha,
            "libro_id": libro["id"],
            "titulo": titulo,
            "precio_original": str(
                libro["precio_original"]
            ),
            "moneda_original": libro["moneda"],
            "cotizacion_oficial": (
                str(libro["cotizacion"])
                if libro["cotizacion"] is not None
                else ""
            ),
            "fecha_cotizacion": (
                str(libro["fecha_cotizacion"])
                if libro["fecha_cotizacion"] is not None
                else ""
            ),
            "precio_interno_ars": (
                str(libro["precio_ars"])
                if libro["precio_ars"] is not None
                else ""
            ),
            "precio_web_ars": "",
            "diferencia_ars": "",
            "diferencia_porcentaje": "",
            "umbral_porcentaje": str(umbral),
            "estado": "",
            "url": "",
        }

        # Validar conversión monetaria.
        if libro["estado_conversion"] != "OK":

            fila = fila_base.copy()

            fila["estado"] = libro["estado_conversion"]

            reporte.append(fila)

            continue

        # Si no encontramos el título exacto.
        if not coincidencias:

            fila = fila_base.copy()

            fila["estado"] = "SIN COINCIDENCIA"

            reporte.append(fila)

            continue

        comparaciones_validas = 0

        for resultado in coincidencias:

            try:

                precio_web = Decimal(
                    str(resultado.get("precio"))
                )

                if (
                    not precio_web.is_finite()
                    or precio_web <= 0
                ):
                    continue

            except (
                InvalidOperation,
                TypeError,
                ValueError,
            ):
                continue

            precio_interno = libro["precio_ars"]

            if (
                precio_interno is None
                or not precio_interno.is_finite()
                or precio_interno <= 0
            ):
                continue

            # Diferencia absoluta en pesos.
            diferencia = (
                precio_web - precio_interno
            )

            # Diferencia porcentual.
            porcentaje = (
                diferencia / precio_interno
            ) * Decimal("100")

            fila = fila_base.copy()

            fila["precio_web_ars"] = str(
                precio_web
            )

            fila["diferencia_ars"] = str(
                diferencia.quantize(
                    Decimal("0.01")
                )
            )

            fila["diferencia_porcentaje"] = (
                f"{porcentaje:.2f}"
            )

            # Se considera tanto aumento como disminución.
            fila["estado"] = (
                "ALERTA"
                if abs(porcentaje) >= umbral
                else "SIN ALERTA"
            )

            fila["url"] = resultado.get(
                "url",
                ""
            )

            reporte.append(fila)

            comparaciones_validas += 1

        if comparaciones_validas == 0:

            fila = fila_base.copy()

            fila["estado"] = (
                "SIN PRECIO WEB VÁLIDO"
            )

            reporte.append(fila)

    return reporte


# ============================================================
# GUARDAR CSV
# ============================================================

def guardar_csv(ruta_csv, reporte):
    """
    Guarda todas las comparaciones en un CSV,
    incluyendo los libros sin alerta o coincidencia.
    """

    ruta_csv.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    with ruta_csv.open(
        "w",
        newline="",
        encoding="utf-8-sig",
    ) as archivo:

        escritor = csv.DictWriter(
            archivo,
            fieldnames=COLUMNAS_CSV,
            delimiter=";",
        )

        escritor.writeheader()

        escritor.writerows(reporte)


# ============================================================
# PROCESO PRINCIPAL
# ============================================================

def main():
    """
    Ejecuta el proceso completo del Ejercicio 04.
    """

    proyecto = Path(__file__).resolve().parents[3]

    print("=== BOOK MANAGER - EJERCICIO 04 ===")

    # Solicitar porcentaje antes de ejecutar el Spider.
    umbral = solicitar_umbral()

    # Conectar con Supabase.
    conexion = ConexionDB()

    # Consultar catálogo.
    catalogo = obtener_catalogo(conexion)

    if not catalogo:

        raise ValueError(
            "No hay libros con precios para comparar."
        )

    # Obtener última cotización oficial.
    dolar_oficial = obtener_dolar_oficial(
        conexion
    )

    # Convertir precios a ARS.
    convertir_precios(
        catalogo,
        dolar_oficial
    )

    if dolar_oficial:

        print(
            "\nDólar oficial venta:",
            dolar_oficial["venta"],
            "ARS",
        )

        print(
            "Fecha de cotización:",
            dolar_oficial["fecha"],
        )

    else:

        print(
            "\nNo se encontró una cotización oficial válida."
        )

    # Títulos únicos del catálogo.
    titulos = list(
        dict.fromkeys(
            libro["titulo"].strip()
            for libro in catalogo
            if (
                libro["titulo"]
                and libro["titulo"].strip()
            )
        )
    )

    # Identificador único de ejecución.
    fecha_archivo = datetime.now().strftime(
        "%Y%m%d_%H%M%S_%f"
    )

    fecha_reporte = datetime.now().isoformat(
        timespec="seconds"
    )

    # Carpeta de reportes.
    carpeta_reportes = (
        proyecto / "src/book_manager/reportes"
    )

    carpeta_reportes.mkdir(
        parents=True,
        exist_ok=True,
    )

    # JSON intermedio del scraping.
    ruta_json = (
        carpeta_reportes
        / f"scraping_ejercicio04_{fecha_archivo}.json"
    )

    # CSV final de alertas.
    ruta_csv = (
        carpeta_reportes
        / f"alertas_precios_{fecha_archivo}.csv"
    )

    print(
        "\nLibros del catálogo:",
        len(titulos)
    )

    print(
        "Porcentaje de alerta:",
        f"{umbral}%"
    )

    # Ejecutar Spider.
    resultados_web = ejecutar_spider(
        proyecto,
        titulos,
        ruta_json,
    )

    # Comparar precios.
    reporte = generar_reporte(
        catalogo,
        resultados_web,
        umbral,
        fecha_reporte,
    )

    # Generar CSV.
    guardar_csv(ruta_csv,reporte )

    # Resumen de estados.
    estados = {}

    for fila in reporte:

        estado = fila["estado"]

        estados[estado] = (
            estados.get(estado, 0) + 1
        )

    print("\n=== RESUMEN ===")

    print(
        "Resultados web:",
        len(resultados_web)
    )

    print(
        "Filas del reporte:",
        len(reporte)
    )

    for estado, cantidad in sorted(
        estados.items()
    ):

        print(
            f"{estado}: {cantidad}"
        )

    print(
        "\nArchivo CSV:",
        ruta_csv
    )

    return ruta_csv


# ============================================================
# EJECUCIÓN DIRECTA DEL MÓDULO
# ============================================================

if __name__ == "__main__":
    main()

'''


# ============================================================
# 3. CREAR FÍSICAMENTE ejecutar_scraper.py
# ============================================================

compile(codigo, str(RUTA_MODULO),"exec")

RUTA_MODULO.parent.mkdir(parents=True,exist_ok=True)
RUTA_MODULO.write_text(codigo,encoding="utf-8")

# ============================================================
# 4. CONFIGURAR SUPABASE Y CORREGIR RUTAS DE IMPORTACIÓN
# ============================================================

os.environ["DATABASE_URL"] = database_url

# Priorizar la carpeta src del proyecto.
if str(SRC) in sys.path:
    sys.path.remove(str(SRC))

sys.path.insert(0, str(SRC))

# Eliminar módulos anteriores de book_manager que puedan
# haberse cargado desde otra ubicación en Colab.
for nombre in list(sys.modules):
    if nombre == "book_manager" or nombre.startswith("book_manager."):
        del sys.modules[nombre]

importlib.invalidate_caches()

# ============================================================
# 5. EJECUTAR EL SCRAPER Y GENERAR CSV
# ============================================================

from book_manager.scraper.ejecutar_scraper import main

ruta_csv = main()

print("\n=== CSV GENERADO ===")
print("Archivo:", ruta_csv)
print("¿Existe?:", ruta_csv.is_file())
````

````python
# @title Reporte en Excel

import sys
import importlib
from pathlib import Path

# ============================================================
# 1. CONFIGURACIÓN DEL PROYECTO
# ============================================================

PROYECTO = Path(
    "/content/Seminario_de_Actualizacion_1/book_manager"
)

SRC = PROYECTO / "src"

RUTA_MODULO = (
    SRC / "book_manager/reportes/reporte_excel.py"
)


# ============================================================
# 2. CÓDIGO DEL MÓDULO reporte_excel.py
# ============================================================

codigo = r'''
"""
Ejercicio 05 - Book Manager.

Genera un reporte Excel a partir del CSV del Ejercicio 04.

Columnas:
- Libro
- Precio interno
- Precio web
- Diferencia
- Fecha de extracción
"""

import csv
from datetime import datetime
from decimal import Decimal, InvalidOperation
from pathlib import Path

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment


# ============================================================
# CONVERTIR VALORES NUMÉRICOS
# ============================================================

def convertir_numero(valor):
    """
    Convierte un valor del CSV a número.

    Si está vacío o no es válido, devuelve None.
    """

    if valor is None or str(valor).strip() == "":
        return None

    try:
        numero = Decimal(str(valor).strip())

        if not numero.is_finite():
            return None

        return float(numero)

    except (InvalidOperation, ValueError, TypeError):
        return None


# ============================================================
# CONVERTIR FECHA
# ============================================================

def convertir_fecha(valor):
    """
    Convierte la fecha de ejecución del CSV
    a un valor de fecha compatible con Excel.
    """

    if not valor:
        return None

    try:
        return datetime.fromisoformat(valor)

    except (ValueError, TypeError):
        return None


# ============================================================
# LEER CSV DEL EJERCICIO 04
# ============================================================

def leer_csv(ruta_csv):
    """
    Lee el CSV generado por el Ejercicio 04.
    """

    ruta_csv = Path(ruta_csv)

    if not ruta_csv.is_file():
        raise FileNotFoundError(
            f"No se encontró el CSV: {ruta_csv}"
        )

    with ruta_csv.open(
        "r",
        encoding="utf-8-sig",
        newline=""
    ) as archivo:

        lector = csv.DictReader(
            archivo,
            delimiter=";"
        )

        columnas_requeridas = {
            "titulo",
            "precio_interno_ars",
            "precio_web_ars",
            "diferencia_ars",
            "fecha_ejecucion",
        }

        if not columnas_requeridas.issubset(
            set(lector.fieldnames or [])
        ):
            raise ValueError(
                "El CSV no contiene las columnas necesarias."
            )

        return list(lector)


# ============================================================
# GENERAR REPORTE EXCEL
# ============================================================

def generar_excel(ruta_csv, ruta_excel=None):
    """
    Genera un archivo Excel con las cinco columnas
    solicitadas en el Ejercicio 05.

    Cuando no existe precio web:
    - Precio web = Sin datos encontrados
    - Diferencia = Sin datos encontrados
    """

    ruta_csv = Path(ruta_csv)

    registros = leer_csv(ruta_csv)

    if ruta_excel is None:
        ruta_excel = ruta_csv.with_name(
            ruta_csv.stem.replace(
                "alertas_precios",
                "reporte_precios"
            ) + ".xlsx"
        )

    ruta_excel = Path(ruta_excel)

    ruta_excel.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    # Crear libro de Excel
    libro_excel = Workbook()

    hoja = libro_excel.active
    hoja.title = "Comparación de precios"

    # Encabezados solicitados
    encabezados = [
        "Libro",
        "Precio interno",
        "Precio web",
        "Diferencia",
        "Fecha de extracción",
    ]

    hoja.append(encabezados)

    # ========================================================
    # CARGAR REGISTROS
    # ========================================================

    for registro in registros:

        titulo = registro["titulo"]

        precio_interno = convertir_numero(
            registro["precio_interno_ars"]
        )

        precio_web = convertir_numero(
            registro["precio_web_ars"]
        )

        diferencia = convertir_numero(
            registro["diferencia_ars"]
        )

        # Cuando no se encontró un precio válido en la web,
        # mostrar un mensaje en lugar de cero.
        if precio_web is None:

            precio_web = "Sin datos encontrados"
            diferencia = "Sin datos encontrados"

        fecha = convertir_fecha(
            registro["fecha_ejecucion"]
        )

        hoja.append([
            titulo,
            precio_interno,
            precio_web,
            diferencia,
            fecha,
        ])

    # ========================================================
    # FORMATO DE ENCABEZADOS
    # ========================================================

    color_encabezado = PatternFill(
        fill_type="solid",
        fgColor="17365D"
    )

    fuente_encabezado = Font(
        color="FFFFFF",
        bold=True
    )

    for celda in hoja[1]:

        celda.fill = color_encabezado
        celda.font = fuente_encabezado

        celda.alignment = Alignment(
            horizontal="center"
        )

    # ========================================================
    # FORMATO DE VALORES
    # ========================================================

    for fila in hoja.iter_rows(
        min_row=2,
        max_row=hoja.max_row
    ):

        # Precio interno, web y diferencia
        for celda in fila[1:4]:

            # Aplicar formato moneda únicamente a números.
            if isinstance(celda.value, (int, float)):

                celda.number_format = (
                    '"$" #,##0.00;[Red]("$" #,##0.00)'
                )

        # Fecha de extracción
        fila[4].number_format = "DD/MM/YYYY HH:MM"

    # ========================================================
    # ANCHO DE COLUMNAS
    # ========================================================

    anchos = {
        "A": 40,
        "B": 20,
        "C": 25,
        "D": 25,
        "E": 24,
    }

    for columna, ancho in anchos.items():
        hoja.column_dimensions[columna].width = ancho

    # Filtros
    hoja.auto_filter.ref = hoja.dimensions

    # Mantener encabezados visibles
    hoja.freeze_panes = "A2"

    # ========================================================
    # GUARDAR EXCEL
    # ========================================================

    libro_excel.save(ruta_excel)

    print("\n=== REPORTE EXCEL GENERADO ===")
    print("Registros:", len(registros))
    print("Archivo:", ruta_excel)

    return ruta_excel


# ============================================================
# EJECUCIÓN DIRECTA
# ============================================================

if __name__ == "__main__":
    print(
        "Este módulo debe ejecutarse indicando "
        "el CSV del Ejercicio 04."
    )

'''


# ============================================================
# 3. VALIDAR Y GUARDAR EL MÓDULO
# ============================================================

compile(
    codigo,
    str(RUTA_MODULO),
    "exec"
)

RUTA_MODULO.parent.mkdir(
    parents=True,
    exist_ok=True
)

RUTA_MODULO.write_text(
    codigo,
    encoding="utf-8"
)


# ============================================================
# 4. CONFIGURAR IMPORTACIONES DE COLAB
# ============================================================

if str(SRC) in sys.path:
    sys.path.remove(str(SRC))

sys.path.insert(0, str(SRC))

for nombre in list(sys.modules):
    if nombre == "book_manager" or nombre.startswith("book_manager."):
        del sys.modules[nombre]

importlib.invalidate_caches()
````

````python
# @title Creacion decorador de auditoría

from pathlib import Path

BASE = Path(
    "/content/Seminario_de_Actualizacion_1/"
    "book_manager/src/book_manager"
)

CARPETA = BASE / "auditoria"
CARPETA.mkdir(parents=True, exist_ok=True)

(CARPETA / "__init__.py").touch()

RUTA_AUDITORIA = CARPETA / "auditoria.py"

codigo = '''
"""
Decorador para auditar operaciones de Book Manager.

Registra la acción, fecha y detalles de cada operación
en la tabla auditoria de la base de datos.
"""

from datetime import datetime
from functools import wraps

from book_manager.database.connection import ConexionDB
from book_manager.models.models import AuditoriaModel


def registrar_auditoria(accion: str, detalles: str) -> None:
    """Guarda un registro de auditoría en la base de datos."""

    conexion = ConexionDB()

    with conexion.obtener_sesion() as sesion:
        registro = AuditoriaModel(
            accion=accion,
            fecha=datetime.now(),
            detalles=detalles,
        )

        sesion.add(registro)


def auditar(accion: str):
    """
    Decorador que registra el resultado de una operación.

    Si la operación falla, registra el error y vuelve
    a lanzar la excepción original.
    """

    def decorador(funcion):

        @wraps(funcion)
        def envoltura(*args, **kwargs):

            try:
                resultado = funcion(*args, **kwargs)

            except Exception as error:

                try:
                    registrar_auditoria(
                        accion=accion,
                        detalles=(
                            f"Error en {funcion.__qualname__}: "
                            f"{type(error).__name__}: {error}"
                        ),
                    )
                except Exception as error_auditoria:
                    print(
                        "No se pudo registrar la auditoría: "
                        f"{error_auditoria}"
                    )

                raise

            registrar_auditoria(
                accion=accion,
                detalles=(
                    f"Operación {funcion.__qualname__} "
                    "completada correctamente."
                ),
            )

            return resultado

        return envoltura

    return decorador
'''

# Validar antes de guardar
compile(codigo, str(RUTA_AUDITORIA), "exec")

RUTA_AUDITORIA.write_text(
    codigo,
    encoding="utf-8"
)

print("=== EJERCICIO 06 ===")
print("Decorador de auditoría creado correctamente.")
print("Archivo:", RUTA_AUDITORIA)
````

````python
# @title Incorporacion del decorador a los servicios

from pathlib import Path
import ast

RUTA = Path(
    "/content/Seminario_de_Actualizacion_1/"
    "book_manager/src/book_manager/services/services.py"
)

codigo = RUTA.read_text(encoding="utf-8")

# Importar el decorador
importacion = "from book_manager.auditoria.auditoria import auditar"

if importacion not in codigo:
    codigo = codigo.replace(
        "from book_manager.models.models import (",
        importacion + "\n\nfrom book_manager.models.models import (",
        1
    )

# Acciones a auditar por clase y método
operaciones = {
    "ServicioBase": {
        "crear": "CREAR",
        "obtener_por_id": "CONSULTAR_POR_ID",
        "obtener_todos": "LISTAR",
        "actualizar": "ACTUALIZAR",
        "eliminar": "ELIMINAR",
    },
    "ServicioLibro": {
        "crear": "CREAR_LIBRO",
    },
    "ServicioPrecio": {
        "crear": "CREAR_PRECIO",
    },
    "ServicioStock": {
        "listar": "LISTAR_STOCK",
        "crear": "CREAR_STOCK",
        "obtener_por_libro": "CONSULTAR_STOCK",
        "actualizar": "ACTUALIZAR_STOCK",
        "eliminar": "ELIMINAR_STOCK",
    },
    "ServicioCotizacionDolar": {
        "crear": "CREAR_COTIZACION",
        "obtener_por_tipo_y_fecha": "CONSULTAR_COTIZACION",
        "obtener_historico": "CONSULTAR_HISTORICO_COTIZACION",
        "actualizar": "ACTUALIZAR_COTIZACION",
        "eliminar": "ELIMINAR_COTIZACION",
        "obtener_cotizaciones": "OBTENER_COTIZACIONES_API",
    },
}

# Analizar el archivo para identificar los métodos reales
arbol = ast.parse(codigo)
lineas = codigo.splitlines(keepends=True)
inserciones = []

for nodo in arbol.body:

    if not isinstance(nodo, ast.ClassDef):
        continue

    if nodo.name not in operaciones:
        continue

    for metodo in nodo.body:

        if not isinstance(
            metodo,
            (ast.FunctionDef, ast.AsyncFunctionDef)
        ):
            continue

        accion = operaciones[nodo.name].get(metodo.name)

        if accion is None:
            continue

        # Evitar agregar dos veces el decorador
        ya_auditado = any(
            isinstance(decorador, ast.Call)
            and isinstance(decorador.func, ast.Name)
            and decorador.func.id == "auditar"
            for decorador in metodo.decorator_list
        )

        if ya_auditado:
            continue

        # Insertar antes del primer decorador existente
        numero_linea = (
            metodo.decorator_list[0].lineno
            if metodo.decorator_list
            else metodo.lineno
        )

        indentacion = " " * metodo.col_offset

        # Identificar el servicio concreto para las operaciones genéricas
        if nodo.name == "ServicioBase":
            accion = f"{accion}_{{type(self).__name__}}"

        if nodo.name == "ServicioBase":
            # Se usará un nombre genérico estable para evitar
            # expresiones dinámicas en los decoradores.
            accion = operaciones[nodo.name][metodo.name]

        inserciones.append(
            (
                numero_linea - 1,
                f'{indentacion}@auditar("{accion}")\n'
            )
        )

# Aplicar las inserciones desde el final del archivo
for posicion, texto in sorted(inserciones, reverse=True):
    lineas.insert(posicion, texto)

codigo_actualizado = "".join(lineas)

# Validar sintaxis
compile(codigo_actualizado, str(RUTA), "exec")

# Guardar cambios
RUTA.write_text(
    codigo_actualizado,
    encoding="utf-8"
)

print("=== EJERCICIO 06 ===")
print("Archivo:", RUTA)
print("Decoradores incorporados:", len(inserciones))
print("Sintaxis validada correctamente.")

print("\n=== OPERACIONES AUDITADAS ===")

for clase, metodos in operaciones.items():
    print(f"\n{clase}:")
    for metodo, accion in metodos.items():
        print(f"  {metodo} -> {accion}")
````

````python
# @title Auditar scraper y reporte Excel

import ast
from pathlib import Path

BASE = Path(
    "/content/Seminario_de_Actualizacion_1/"
    "book_manager/src/book_manager"
)

ARCHIVOS = {
    BASE / "scraper" / "ejecutar_scraper.py": {
        "main": "EJECUTAR_SCRAPER",
    },
    BASE / "reportes" / "reporte_excel.py": {
        "generar_excel": "GENERAR_REPORTE_EXCEL",
    },
}

IMPORTACION = "from book_manager.auditoria.auditoria import auditar"

print("=== EJERCICIO 06 - AUDITORÍA DE OPERACIONES ===")

for ruta, operaciones in ARCHIVOS.items():

    if not ruta.is_file():
        raise FileNotFoundError(f"No existe: {ruta}")

    codigo = ruta.read_text(encoding="utf-8")

    # Agregar importación después de los imports existentes
    if IMPORTACION not in codigo:
        arbol = ast.parse(codigo)

        imports = [
            nodo for nodo in arbol.body
            if isinstance(nodo, (ast.Import, ast.ImportFrom))
        ]

        if not imports:
            raise ValueError(
                f"No se encontraron importaciones en {ruta.name}"
            )

        lineas = codigo.splitlines(keepends=True)
        posicion = max(nodo.end_lineno for nodo in imports)

        lineas.insert(posicion, "\n" + IMPORTACION + "\n")
        codigo = "".join(lineas)

    # Identificar funciones que deben auditarse
    arbol = ast.parse(codigo)
    lineas = codigo.splitlines(keepends=True)

    inserciones = []
    encontradas = set()

    for nodo in arbol.body:

        if not isinstance(
            nodo,
            (ast.FunctionDef, ast.AsyncFunctionDef)
        ):
            continue

        if nodo.name not in operaciones:
            continue

        encontradas.add(nodo.name)

        ya_auditada = any(
            isinstance(decorador, ast.Call)
            and isinstance(decorador.func, ast.Name)
            and decorador.func.id == "auditar"
            for decorador in nodo.decorator_list
        )

        if ya_auditada:
            continue

        posicion = (
            nodo.decorator_list[0].lineno
            if nodo.decorator_list
            else nodo.lineno
        ) - 1

        accion = operaciones[nodo.name]

        inserciones.append(
            (posicion, f'@auditar("{accion}")\n')
        )

    faltantes = set(operaciones) - encontradas

    if faltantes:
        raise ValueError(
            f"Funciones no encontradas en {ruta.name}: {faltantes}"
        )

    # Insertar decoradores desde el final
    for posicion, decorador in sorted(
        inserciones,
        reverse=True
    ):
        lineas.insert(posicion, decorador)

    codigo_actualizado = "".join(lineas)

    # Validar sintaxis antes de guardar
    compile(codigo_actualizado, str(ruta), "exec")

    ruta.write_text(
        codigo_actualizado,
        encoding="utf-8"
    )

    print(f"\nArchivo: {ruta.name}")
    print(f"Decoradores agregados: {len(inserciones)}")

    for funcion, accion in operaciones.items():
        print(f"  {funcion} -> {accion}")

print("\nArchivos actualizados correctamente.")
````
